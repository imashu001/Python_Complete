# Keywords

**Keywords** are reserved words in Python that have special, built-in meanings and purposes for the language interpreter. They form the foundational vocabulary of the Python language.

---

## The Golden Rule of Keywords
You **cannot** use Python keywords as variable names, function names, class names, or any other user-defined identifiers. Doing so will immediately trigger a `SyntaxError`.

```python
# Invalid: 'for' is a reserved keyword
for = 10  # Raises SyntaxError: invalid syntax

## How to View All Keywords Programmatically
Python's built-in keyword module allows you to check all reserved words available in your specific version of Python.

````python
import keyword

# Print the list of all keywords
print(keyword.kwlist)

# Check if a specific word is a keyword
print(keyword.iskeyword("if"))  # Output: True
print(keyword.iskeyword("username"))  # Output: False
````
