# Naming Conventions (PEP 8)

**PEP 8** (Python Enhancement Proposal 8) is the official style guide for Python code. Writing code that looks clean and follows consistent naming conventions makes your code readable, maintainable, and professional.

---

## Why PEP 8 Matters
* **Readability:** Code is read much more often than it is written. Consistent naming lets you instantly recognize what a piece of data is (e.g., whether it's a class, a constant, or a regular variable) just by looking at its casing.
* **Collaboration:** When everyone follows the same guide, teams can switch codebases effortlessly without adjusting to unique personal styles.

---

## Python Naming Conventions Quick Reference

| Type | Convention | Examples |
| :--- | :--- | :--- |
| **Variables** | `snake_case` | `user_age`, `total_amount`, `is_active` |
| **Functions** | `snake_case` | `calculate_tax()`, `get_user_data()` |
| **Methods** | `snake_case` | `save_to_database()`, `to_string()` |
| **Constants** | `ALL_CAPS` | `PI`, `MAX_CONNECTIONS`, `DEFAULT_TIMEOUT` |
| **Classes** | `PascalCase` (CapWords) | `UserAccount`, `ShoppingCart`, `HttpException` |
| **Modules / Files** | `snake_case` (short, lowercase) | `utils.py`, `data_parser.py`, `auth.py` |
| **Packages** | lowercase (no underscores if possible) | `requests`, `sqlalchemy`, `my_package` |

## Special Naming Conventions (Underscores)

Python uses leading and trailing underscores for special meanings or access control:

* **Single Leading Underscore (`_private_var`):** 
  * Signals an internal/private variable or method (weak "internal use" indicator). It tells other developers, *"Hey, this isn't part of the public API, so be careful if you use it outside this class/module."*
* **Single Trailing Underscore (`class_`):** 
  * Used when your preferred name clashes with a Python keyword or built-in function (e.g., naming a variable `class_` or `list_` instead of `class` or `list`).
* **Double Leading Underscore (`__private_attr`):** 
  * Triggers **name mangling** inside classes, changing the attribute name to prevent accidental overriding in subclasses.
* **Dunder / Magic Names (`__init__`, `__str__`):** 
  * Double leading and trailing underscores are reserved for built-in Python magic methods. Never invent your own names with this pattern.

## Best Practices & Anti-Patterns

### 1. Avoid Ambiguous Single-Letter Names
* **Bad:** `l` (looks like 1), `O` or `I` (look like 0 or 1).
* **Good:** Use descriptive words, except for standard loop counters like `i` or `j`.

### 2. Be Descriptive, Not Cryptic
* **Bad:** `usr_cnt = 45`
* **Good:** `user_count = 45`

### 3. Avoid CamelCase for Variables
* **Bad:** `userAge = 25` (This is standard in JavaScript/Java, but violates PEP 8 for Python).
* **Good:** `user_age = 25`
