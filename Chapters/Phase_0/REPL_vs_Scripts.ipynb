# REPL vs Script

| Feature | REPL | Script |
| :--- | :--- | :--- |
| **Full Name** | Read-Eval-Print Loop | A Python program saved in a `.py` file |
| **Execution** | Executes one line at a time | Executes the entire file |
| **Primary Use** | Used for quick testing and learning | Used for real applications |
| **Interactivity** | Interactive | Non-interactive |
| **Persistence** | No need to save code | Code is saved and reusable |

---

## REPL
*(Read-Eval-Print Loop)*

* It is an interactive environment where we write and execute small snippets of code immediately and see the output instantly.
* **Examples:** Python terminal, IPython, Jupyter Notebook cells, VS Code Python console.

---

## Script

* It is a Python program saved in a `.py` file mainly used to write reusable code.
* The complete file is executed as a program, usually triggered from the terminal using commands like: `python filename.py`
