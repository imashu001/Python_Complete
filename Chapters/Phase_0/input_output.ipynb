# Input & Output (I/O)

Interaction with the user is a core part of most programs. In Python, outputting data to the screen and taking input from the user are handled using built-in functions.

---

## 1. Output: The `print()` Function

The `print()` function is used to send data (text, numbers, variables) to the console window.

```python
# Basic printing
print("Hello, World!")

# Printing multiple items (automatically separated by a space)
name = "Alice"
age = 25
print("Name:", name, "Age:", age)

Customizing print()
sep (Separator): Defines what separates multiple arguments (default is a single space " ").

````python
print("2026", "08", "05", sep="-")  # Output: 2026-08-05
````

end (Ending): Defines what is printed at the very end of the line (default is a newline "\n").

````python
print("Loading", end="...")
print("Done!")  # Output: Loading...Done!
````

# 2. Input: The input() Function
The input() function pauses the program, waits for the user to type something into the console, and presses Enter. It takes an optional prompt string as an argument.

````python
username = input("Enter your username: ")
print("Welcome back,", username)
````

⚠️ The input() String-Return Caveat
Crucial Rule: The input() function always returns the user's input as a string (str), no matter what the user types—even if they type a pure number.

If you try to perform mathematical operations directly on raw input without converting its type, your program will either throw a TypeError or produce unexpected results (string concatenation instead of addition).

Problem :

````python
# User enters: 10
age = input("Enter your age: ")

# Trying to add 5 to age
# future_age = age + 5  -> Raises TypeError: can only concatenate str (not "int") to str
````

If you join two strings:

````python
num1 = input("Enter first number: ")  # User types: 5
num2 = input("Enter second number: ")  # User types: 5

print(num1 + num2)  # Output: 55 (String concatenation, NOT 10!)
````

The Solution (Type Casting):
To use user input in mathematical calculations, you must explicitly convert the returned string into the desired data type using functions like int() or float().

````python
# Correct way to handle numeric input
num1 = int(input("Enter first number: "))  # Converted to integer
num2 = int(input("Enter second number: "))  # Converted to integer

print(num1 + num2)  # Output: 10 (Correct mathematical addition!)
````
