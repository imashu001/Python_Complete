# Built-In Functions

**Built-in functions** are pre-written, standard functions provided directly by Python. They are always available for you to use in any script or REPL environment without needing to import any modules or libraries.

---

## What is a Built-in Function?
Think of built-in functions as tools already built into your Python toolbox. You simply call them by their name followed by parentheses `()`, passing in any required data (arguments).

```python
# Example of using a built-in function
length = len("Hello, World!")
print(length)  # Output: 13

## Essential Built-in Functions for Beginners
Here is a curated list of the most common built-in functions you will use daily:

### 1. Introspection & Data Inspection
type(object): Returns the data type of an object.

````python
print(type(42))  # Output: <class 'int'>
print(type("Python"))  # Output: <class 'str'>
````

len(object): Returns the length (number of items) of a string, list, dictionary, or other collection.

````python
print(len([10, 20, 30, 40]))  # Output: 4
````

id(object): Returns the unique memory address identifier of an object.

````python
x = 10
print(id(x))  # Output: (unique memory integer)
````

### 2. Type Conversion
int(), float(), str(), bool(): Convert values from one data type to another

````python
num_str = "100"
num_int = int(num_str)  # Converts string to integer
print(type(num_int))  # Output: <class 'int'>
````

### 3. Math & Numeric Operations
abs(number): Returns the absolute (positive) value of a number.

````python
print(abs(-15))  # Output: 15
````

round(number, digits): Rounds a floating-point number to a specified number of decimal places.

````python
print(round(3.14159, 2))  # Output: 3.14
````

min() & max(): Returns the smallest or largest item in an iterable or a set of arguments.

````python
print(max(5, 12, 3))  # Output: 12
````

sum(iterable): Adds up all the elements in a numeric collection.

````python
print(sum([1, 2, 3, 4]))  # Output: 10
````

#### 4. Sequence Generation
range(start, stop, step): Generates a sequence of numbers, commonly used in loops.

````python
print(list(range(1, 6)))  # Output: [1, 2, 3, 4, 5]
````

# How to View All Built-in Functions
Python has dozens of built-in functions. You can programmatically list all of them by inspecting the __builtins__ module namespace:

````python
import builtins

# Prints a list of all built-in functions and exceptions
print(dir(builtins))
````
