# Variables & Assignment

A **variable** is a named reference to a value stored in the computer's memory. Think of it as a labeled box where you can store data to use and manipulate later in your program.

---

## Variable Assignment (`=`)

In Python, variables are created the moment you first assign a value to them using the assignment operator (`=`). Unlike statically typed languages (like C or Java), you do not need to declare a variable's data type beforehand.

```python
x = 10  # Integer
name = "Alice"  # String
price = 19.99  # Float
is_active = True  # Boolean
```

Unlike languages where variables are boxes containing values, Python variables are labels or pointers pointing to objects in memory.

```python
# x points to the integer object 50
x = 50

# x now points to a completely new integer object 100
# The original object 50 is left for garbage collection if no other variables point to it
x = 100
```

```python
#1. Assigning the same value to multiple variables:
x = y = z = 0

#2. Assigning multiple values to multiple variables:
a, b, c = 1, 2, "Hello"
```

# Constants by Convention
Python does not have built-in support for true immutable constants (like const in JavaScript or C++). Instead, the community uses ALL_CAPS as a naming convention to signal to other developers that a variable should not be changed.
