# Comments

Comments are text notes written inside your Python code for humans to read. Python completely ignores comments when executing the program; they are strictly meant for documentation, explanations, and leaving notes for yourself or other developers.

---

## Types of Comments in Python

Python supports two primary ways to write comments:

### 1. Single-Line Comments (`#`)
Anything following a hash symbol (`#`) on that line is treated as a comment.

```python
# This is a full-line comment explaining the next block of code
print("Hello, World!")  # This is an inline comment explaining this specific line
```

### 2. Multi-Line Comments (Triple Quotes)

While Python does not have a dedicated multi-line comment syntax, the convention is to use triple quotes (""" or '''). Technically, these are string literals (often used as docstrings), but if they aren't assigned to a variable, Python ignores them at runtime.

```python
"""
This is a multi-line comment.
It can span across multiple lines and is typically
used for longer explanations or temporarily disabling code blocks.
"""
print("Python is running!")
```

```python
x = 0
# Add 5 to x (bad comment)
x = x + 5

# **Good Comment (Contextual):**
# Add a buffer to account for network latency spikes
x = x + 5
```

### 2. Keep Comments Updated
Outdated comments are worse than no comments at all because they actively mislead anyone reading the code. When you change your code, remember to update or remove the accompanying comments.

### 3. Avoid Obvious Comments
Don't clutter your codebase with comments that repeat the obvious syntax.

Bad: name = "Alice"  # Sets name to Alice

Good: No comment needed here; the variable name is self-explanatory.

# Docstrings (Documentation Strings)
A special kind of comment used specifically to document modules, classes, functions, and methods. They are placed as the very first statement inside a block and can be accessed programmatically using the __doc__ attribute or tools like help().

```python
def calculate_area(length, width):
  """Calculates the area of a rectangle.

  Args:
      length (float): The length of the rectangle.
      width (float): The width of the rectangle.

  Returns:
      float: The calculated area.
  """
  return length * width


# Accessing the docstring
print(calculate_area.__doc__)
```
