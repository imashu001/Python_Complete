# Indentation

Unlike most programming languages that use curly braces `{}` or keywords (like `begin` and `end`) to define code blocks, Python uses **whitespace and indentation**. 

Indentation is not just a matter of style in Python—it is a **strict syntactic requirement**.

---

## Why Indentation Matters
Python uses indentation to determine the **scope and hierarchy** of code. Lines of code that share the same level of indentation belong to the same code block (such as inside a loop, function, or conditional statement).

```python
# Example of correct indentation
if True:
    print("This line is inside the if block")
    print("This line is also inside the if block")

print("This line is outside the if block")

## The Golden Rules of Indentation
### 1. Follow PEP 8 (Use 4 Spaces)
The official Python style guide (PEP 8) dictates that you should use 4 spaces per indentation level.

Good: Using 4 spaces using your text editor's Tab key (most modern editors automatically convert tabs to 4 spaces).

Bad: Inconsistent spacing (e.g., indenting one line with 2 spaces and the next with 4).

### 2. Never Mix Tabs and Spaces
Python strictly prohibits mixing tabs and spaces for indentation. Doing so will immediately trigger a TabError when you run your script. Stick strictly to spaces.

## Common Errors Related to Indentation
### 1. IndentationError
This occurs when you forget to indent a block of code where Python expects one (such as right after an if statement or a function definition).

````python
# Error: Expected an indented block
if True:
print("Hello")  # Raises IndentationError
````

### 2. TabError
This occurs when your file contains a confusing mix of actual tab characters and space characters.

````python
if True:
    print("Space indented")
	    print("Tab indented")  # Raises TabError: inconsistent use of tabs and spaces in indentation
````
