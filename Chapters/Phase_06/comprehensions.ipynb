# 1. List Comprehensions
Instead of using a traditional for loop and the .append() method to build a list, a list comprehension streamlines the process into a compact syntax.

### Basic Syntax
[expression for item in iterable]

### Adding Conditions (if filtering)
You can filter items by appending an if condition at the end.

### Conditional Expressions (if-else)
If you want to modify values based on a condition, the if-else block moves to the left of the for loop (acting as an inline ternary operator).

```python
# Traditional loop vs. List Comprehension
# Traditional:
squares_trad = []
for x in range(10):
  squares_trad.append(x**2)

# Comprehension:
squares = [x**2 for x in range(10)]

# Filtering with 'if' (Even squares only)
even_squares = [x**2 for x in range(10) if x % 2 == 0]

# Ternary operator in comprehension (Label even/odd)
labels = ["Even" if x % 2 == 0 else "Odd" for x in range(5)]
# Output: ['Even', 'Odd', 'Even', 'Odd', 'Even']
```

# 2. Dictionary & Set Comprehensions
Just like lists, Python supports comprehensions for dictionaries and sets, using curly braces {}.

## Set Comprehensions
Set comprehensions use the same syntax as list comprehensions, but automatically discard duplicate values.

## Dictionary Comprehensions
Dictionary comprehensions require a key-value pair separated by a colon (key: value) in the expression portion.

```python
# Set Comprehension: Unique lengths of words
words = ["apple", "banana", "cherry", "date", "fig"]
lengths = {len(w) for w in words}
print(lengths)  # Output: {3, 4, 5, 6} (unordered)

# Dictionary Comprehension: Map numbers to their squares
num_dict = {x: x**2 for x in range(5)}
print(num_dict)  # Output: {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}

# Swapping keys and values in an existing dictionary
original = {"a": 1, "b": 2, "c": 3}
swapped = {v: k for k, v in original.items()}
print(swapped)  # Output: {1: 'a', 2: 'b', 3: 'c'}
```

# 3. Generator Expressions
Generator expressions look almost identical to list comprehensions, but they use parentheses () instead of brackets [] or curly braces {}.

### The Power of Lazy Evaluation
**List Comprehensions** compute and store every single item in memory all at once.

**Generator Expressions use lazy evaluation** —they generate items on-the-fly one by one only when asked. This makes them extremely memory-efficient for large datasets or infinite streams.

```python
# List comprehension (allocates memory for 1 million items immediately)
list_comp = [x**2 for x in range(1_000_000)]

# Generator expression (memory efficient, evaluates on demand)
gen_expr = (x**2 for x in range(1_000_000))

print(next(gen_expr))  # Output: 0
print(next(gen_expr))  # Output: 1
# You can also iterate through them in loops without loading everything into RAM
```

# 4. Nested Comprehensions
You can nest multiple loops inside a single comprehension to handle multi-dimensional data, such as flattening a matrix or generating coordinate pairs.

**Readability Warning:** While nested comprehensions are powerful, nesting them more than two levels deep can quickly make code unreadable. If it hurts clarity, opt for a traditional multi-line for loop instead.

```python
# Flattening a matrix (List of lists) into a single list
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

flattened = [item for row in matrix for item in row]
print(flattened)  # Output: [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Generating coordinate pairs (Cartesian product)
coordinates = [(x, y) for x in range(3) for y in range(3)]
print(coordinates)
# Output: [(0, 0), (0, 1), (0, 2), (1, 0), (1, 1), (1, 2), (2, 0), (2, 1), (2, 2)]
```
