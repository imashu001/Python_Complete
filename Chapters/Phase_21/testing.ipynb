# Testing
Automated testing is a hallmark of professional software engineering. It ensures your code works as expected, prevents regressions when you refactor, and serves as living documentation. Python has a rich ecosystem for writing and running tests.

## 1. unittest
The unittest module is Python’s built-in testing framework. Inspired by Java’s JUnit, it is object-oriented, requiring you to write test cases as classes that inherit from unittest.TestCase.

```python
import unittest

# Code to test
def add(a, b):
    return a + b

# Test Case Class
class TestMathOperations(unittest.TestCase):
    
    def test_add_positive_numbers(self):
        result = add(2, 3)
        self.assertEqual(result, 5)

    def test_add_negative_numbers(self):
        result = add(-1, -1)
        self.assertEqual(result, -2)

if __name__ == "__main__":
    unittest.main()
```

## 2. pytest
While unittest is built-in, pytest is the industry standard third-party testing framework for Python. It requires far less boilerplate code, uses standard Python assert statements, and features powerful auto-discovery.

```python
# test_math.py
def add(a, b):
    return a + b

# Pytest finds functions starting with 'test_' automatically
def test_add():
    assert add(2, 3) == 5
    assert add(-1, 1) == 0
```

Run it via the terminal using simply: pytest

## 3. Fixtures (pytest Fixtures)
Fixtures are functions executed by pytest before (and optionally after) your actual test functions. They are used to set up a fixed state, such as database connections, dummy configuration files, or user sessions.

```python
import pytest

@pytest.fixture
def sample_user_data():
    # Setup phase
    print("\nSetting up user data...")
    data = {"username": "alice", "role": "admin"}
    yield data  # This provides the data to the test function
    # Teardown phase (runs after the test completes)
    print("\nTearing down user data...")

def test_user_role(sample_user_data):
    assert sample_user_data["role"] == "admin"
```

## 4. Parameterized Tests
Instead of writing repetitive test functions for different inputs, parameterization lets you run a single test function multiple times with different datasets.

```python
import pytest

def is_even(number):
    return number % 2 == 0

# Test multiple inputs using @pytest.mark.parametrize
@pytest.mark.parametrize("input_num, expected", [
    (2, True),
    (3, False),
    (0, True),
    (-4, True)
])
def test_is_even(input_num, expected):
    assert is_even(input_num) == expected
```

## 5. Mocking
When testing code that interacts with external systems (like sending an email, making a network request, or querying a database), you use mocks to isolate your code. Mocking replaces real dependencies with fake objects whose behavior you control.

Python provides unittest.mock (often used alongside pytest):

```python
from unittest.mock import patch
import requests

# External API function we want to test
def get_user_profile(user_id):
    response = requests.get(f"https://api.example.com/users/{user_id}")
    return response.json()

# Test using @patch to mock the 'requests.get' call
@patch("requests.get")
def test_get_user_profile(mock_get):
    # Configure the mock return value
    mock_get.return_value.json.return_value = {"id": 1, "name": "Alice"}

    result = get_user_profile(1)

    assert result["name"] == "Alice"
    mock_get.assert_called_once_with("https://api.example.com/users/1")
```

## 6. Test Coverage
Test coverage measures the percentage of your source code executed by your test suite. The pytest-cov plugin is the standard tool for measuring this in Python.

Install coverage: pip install pytest-cov

Run tests with coverage reporting:

```python
pytest --cov=my_module --cov-report=term-missing
```

This generates a terminal report showing precisely which lines of code were not tested (missing).

### Best Practices & Common Pitfalls
***Follow the AAA Pattern:*** Structure your test functions clearly into Arrange (set up inputs and state), Act (execute the code being tested), and Assert (verify the results).

***Keep Tests Independent:*** Tests should never depend on the execution order or side effects of other tests. Each test must be able to run in isolation and yield the exact same result.

***Don't Mock Everything:*** Only mock external boundaries (databases, APIs, system clocks, file systems). Mocking your own internal utility functions makes tests brittle and tightly coupled to implementation details.
