# Phase 1: Data Types

Every value in Python has a specific data type. Because Python is dynamically typed, you don't need to explicitly declare a variable's type when you create it; Python automatically determines it based on the assigned value.

---

## 1. Numeric Types

Python supports three distinct numeric types for handling integers, decimals, and complex numbers.

### `int` (Integers)
Whole numbers without a decimal point, of unlimited magnitude (limited only by available memory).
```python
age = 25
population = 8_000_000_000  # Underscores can be used as visual separators
print(type(age))  # Output: <class 'int'>

float (Floating-Point Numbers)
Numbers containing a decimal point or written in scientific notation.

````python
price = 19.99
scientific_notation = 1.5e3  # Equivalent to 1500.0
print(type(price))  # Output: <class 'float'>
````

complex (Complex Numbers)
Numbers with a real and an imaginary part, written using j or J for the imaginary component.

````python
z = 3 + 5j
print(z.real)  # Output: 3.0
print(z.imag)  # Output: 5.0
print(type(z))  # Output: <class 'complex'>
````

## 2. Boolean Type (bool)
Represents truth values used in logic and conditions. There are only two boolean literals in Python: True and False (which are subclasses of integers, where True equals 1 and False equals 0)

````python
is_active = True
is_logged_in = False
print(type(is_active))  # Output: <class 'bool'>

# Booleans in math
print(True + True)  # Output: 2
````

## 3. The None Type (None)
None is a special constant in Python that represents the absence of a value, a null value, or an empty state. It has its own data type called NoneType.

````python
result = None
print(type(result))  # Output: <class 'NoneType'>
````

## 4. Sequence Types
Sequence types are ordered collections of items.

Strings (str)
Immutable sequences of Unicode characters enclosed in single, double, or triple quotes.

````python
name = "Python"
greeting = 'Hello'
multi_line = """This is a
multi-line string."""
````

Lists (list)
Ordered, mutable (changeable) collections that can hold items of mixed data types. Defined using square brackets [].

````python
fruits = ["apple", "banana", "cherry"]
mixed_list = [1, "Hello", 3.14, True]

# Modifying a list (mutable)
fruits[0] = "orange"
````

Tuples (tuple)
Ordered, immutable (unchangeable) collections. Defined using parentheses (). Once created, you cannot add, remove, or modify items.

````python
coordinates = (10.0, 20.0)
single_item_tuple = (42,)  # Note the trailing comma for single-item tuples
````

### 5. Set Types (set)
An unordered collection of unique items. Duplicate elements are automatically removed. Sets are mutable, but their elements must be hashable (immutable). Defined using curly braces {} or set().

````python
unique_numbers = {1, 2, 3, 3, 4}
print(unique_numbers)  # Output: {1, 2, 3, 4} (Duplicates removed)

empty_set = set()  # Correct way to create an empty set ({} creates an empty dict)
````

### 6. Mapping Types (Dictionaries - dict)
Collections of key-value pairs. Keys must be unique and immutable (strings, numbers, tuples), while values can be of any data type. Defined using curly braces with colons {key: value}.

````python
user = {"username": "alice123", "email": "alice@example.com", "age": 30}

# Accessing a value by key
print(user["username"])  # Output: alice123
````

### 7. Binary Types
Binary types are used to manipulate raw binary data (bytes, byte arrays, and memory views), essential for file I/O, networking, and image processing.

bytes
Immutable sequences of bytes (integers from 0 to 255). Prefixed with b.

````python
byte_data = b"Hello"
print(type(byte_data))  # Output: <class 'bytes'>
````

bytearray
The mutable version of bytes. You can modify individual byte values in-place.

````python
mutable_bytes = bytearray(b"Hello")
mutable_bytes[0] = 74  # ASCII for 'J'
print(mutable_bytes)  # Output: bytearray(b'Jello')
````

memoryview
An object that allows Python code to access the internal buffer of an object that supports the buffer protocol (like bytes and bytearray) without copying the data into memory.

````python
data = bytearray(b"Python")
mv = memoryview(data)
print(mv[0])  # Output: 80 (ASCII value of 'P')
````
