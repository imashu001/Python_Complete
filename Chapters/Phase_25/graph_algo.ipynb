## 1. Sliding Window

The Sliding Window pattern is used to perform operations on a specific window size of a given array or string, reducing time complexity from $O(N^2)$ to $O(N)$. Windows can be of fixed size or dynamic size.

```python
# Sliding Window: Maximum sum of a subarray of size k
def max_sub_array_sum(nums: list[int], k: int) -> int:
    if len(nums) < k:
        return 0
        
    # Compute sum of first window
    window_sum = sum(nums[:k])
    max_sum = window_sum
    
    # Slide the window across the rest of the array
    for i in range(len(nums) - k):
        window_sum = window_sum - nums[i] + nums[i + k]
        max_sum = max(max_sum, window_sum)
        
    return max_sum
```

## 2. Two Pointers

The Two Pointers pattern uses two index variables that move across an array or string—typically starting from opposite ends (moving inward) or from the same direction (fast and slow pointers). It allows you to process linear data structures in linear time $O(N)$ with $O(1)$ auxiliary space.

```python
# Two Pointers: Container With Most Water
def max_area(height: list[int]) -> int:
    left, right = 0, len(height) - 1
    max_water = 0
    
    while left < right:
        width = right - left
        current_water = min(height[left], height[right]) * width
        max_water = max(max_water, current_water)
        
        # Move the pointer pointing to the shorter line inward
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1
            
    return max_water
```

## 3. Prefix Sum

Prefix Sum involves pre-computing cumulative sums of an array. It allows you to calculate the sum of any arbitrary subarray range $[i, j]$ in $O(1)$ time after an $O(N)$ preprocessing phase.

```python
class NumArray:
    def __init__(self, nums: list[int]):
        # Build prefix sum array with a leading 0 for edge case handling
        self.prefix = [0] * (len(nums) + 1)
        for i in range(len(nums)):
            self.prefix[i + 1] = self.prefix[i] + nums[i]

    def sum_range(self, left: int, right: int) -> int:
        # Query sum in O(1) time
        return self.prefix[right + 1] - self.prefix[left]
```

## 4. Graph Algorithms & Topological Sort

Graph algorithms explore relationships and paths between interconnected nodes.

DFS & BFS: Depth-First Search explores paths deeply using recursion or a stack, while Breadth-First Search explores outward layer-by-layer using a queue (deque).

Topological Sort: A linear ordering of vertices in a Directed Acyclic Graph (DAG) such that for every directed edge $u \to v$, vertex $u$ comes before $v$. It is commonly implemented using Kahn's algorithm (in-degree tracking with a queue).

```python
from collections import deque

# Topological Sort using Kahn's Algorithm (BFS-based)
def find_order(num_courses: int, prerequisites: list[list[int]]) -> list[int]:
    adj = {i: [] for i in range(num_courses)}
    in_degree = [0] * num_courses
    
    # Build graph and calculate in-degrees
    for dest, src in prerequisites:
        adj[src].append(dest)
        in_degree[dest] += 1
        
    # Queue nodes with 0 in-degree
    queue = deque([i for i in range(num_courses) if in_degree[i] == 0])
    result = []
    
    while queue:
        node = queue.popleft()
        result.append(node)
        
        for neighbor in adj[node]:
            in_degree[neighbor] -= 1
            if in_degree[neighbor] == 0:
                queue.append(neighbor)
                
    # If result contains all courses, no cycle exists
    return result if len(result) == num_courses else []
```

### Best Practices & Common Pitfalls
**Define Window Boundaries Carefully:** In sliding window problems, make sure your window expansion and contraction logic correctly handles edge cases (like empty arrays or windows larger than the array).

**Beware of Cycles in Graphs:** Always track visited nodes using a set or state array (0 = unvisited, 1 = visiting, 2 = visited) during graph traversals to prevent infinite loops.
