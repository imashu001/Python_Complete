## 1. Binary Search
Binary search is an efficient $O(\log N)$ algorithm used to find an element in a sorted array by repeatedly dividing the search space in half.Beyond searching for exact values, binary search can be adapted to find boundary conditions (e.g., "first bad version" or the minimum capacity to ship packages within $D$ days).

```python
def binary_search(nums: list[int], target: int) -> int:
    left, right = 0, len(nums) - 1
    
    while left <= right:
        # Prevent potential integer overflow in other languages (good practice in Python too)
        mid = left + (right - left) // 2
        
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
            
    return -1
```

## 2. Recursion & Backtracking
**Recursion:** A technique where a function solves a problem by calling a smaller instance of itself. Every recursive function requires a definitive base case to halt execution and prevent infinite recursion.

**Backtracking:** An algorithmic refinement of recursion. It builds candidate solutions incrementally and abandons ("backs up") a path the moment it violates constraints or fails to lead to a valid solution.

```python
# Classic Backtracking: Generate all permutations of a list
def permute(nums: list[int]) -> list[list[int]]:
    result = []
    
    def backtrack(path: list[int], options: list[int]):
        if not options:
            result.append(list(path))
            return
        
        for i in range(len(options)):
            # Choose
            path.append(options[i])
            next_options = options[:i] + options[i+1:]
            
            # Explore
            backtrack(path, next_options)
            
            # Un-choose (Backtrack)
            path.pop()
            
    backtrack([], nums)
    return result
```

## 3. Greedy Algorithms

A greedy algorithm makes the locally optimal choice at each decision point with the objective of finding the global optimum.

**Characteristics:** They are extremely fast and efficient (often $O(N \log N)$ due to sorting), but they do not work for all problems.

**Classic Examples:** Activity Selection, Fractional Knapsack, Huffman Coding, and Dijkstra's Shortest Path Algorithm.

```python
# Greedy Example: Assign Cookies to Children
def find_content_children(g: list[int], s: list[int]) -> int:
    g.sort()  # Children greed factors
    s.sort()  # Cookie sizes
    
    child_ptr = 0
    cookie_ptr = 0
    
    while child_ptr < len(g) and cookie_ptr < len(s):
        if s[cookie_ptr] >= g[child_ptr]:
            child_ptr += 1  # Child is satisfied, move to next child
        cookie_ptr += 1     # Always consume the cookie
        
    return child_ptr
```

## 4. Dynamic Programming (DP)
Dynamic Programming solves complex problems by breaking them down into simpler, overlapping subproblems and storing their solutions to avoid redundant calculations.

**Top-Down (Memoization):** A recursive approach where results of subproblems are cached (e.g., using @functools.lru_cache or a dictionary).

**Bottom-Up (Tabulation):** An iterative approach that builds solutions from base cases upward using a DP table (array).

```python
# Dynamic Programming: Fibonacci (Bottom-Up Tabulation)
def fibonacci(n: int) -> int:
    if n <= 1:
        return n
        
    dp = [0] * (n + 1)
    dp[1] = 1
    
    for i in range(2, n + 1):
        dp[i] = dp[i-1] + dp[i-2]
        
    return dp[n]
```

### Best Practices & Common Pitfalls
**Check Your Base Cases:** In recursion and backtracking, missing or incorrect base cases will trigger a RecursionError due to stack overflow.

**Verify Greedy Choice Property:** Never use a greedy approach unless you can mathematically prove that local choices lead to a global optimum. For optimization problems with overlapping constraints, Dynamic Programming is safer.
