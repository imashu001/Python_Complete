# Logging
Professional applications don't use print() statements to track execution or debug errors. Instead, they use structured logging. The built-in logging module is powerful, flexible, and essential for monitoring application health in production.

## 1. The logging Module
The logging module allows you to record messages with specific importance levels, configure where those messages go (console, files, external services), and format them with timestamps and module names.

```python
import logging

# Basic configuration
logging.basicConfig(level=logging.DEBUG)

logging.debug("This is a detailed debug message.")
logging.info("Application initialized successfully.")
logging.warning("Disk space is running low.")
logging.error("Failed to connect to the database.")
logging.critical("System failure! Shutting down immediately.")
```

## 2. Log Levels
Logging levels allow you to filter messages based on severity. Python defines 5 standard levels (from lowest to highest severity):

DEBUG: Detailed information, typically of interest only when diagnosing problems.

INFO: Confirmation that things are working as expected.

WARNING: An indication that something unexpected happened (or is about to happen), but the software is still working as expected.

ERROR: A more serious problem that prevented the software from performing a function.

CRITICAL: A serious error indicating that the program itself may be unable to continue running.

Note: When you set a logging level (e.g., INFO), all messages below that level (DEBUG) are ignored.

## 3. Formatters
Formatters define the layout of the log record. You can include timestamps, log levels, thread names, file names, and the actual log message using standard log record attributes.

```python
import logging

# Create a custom logger
logger = logging.getLogger("MyApplication")
logger.setLevel(logging.INFO)

# Create a console handler
handler = logging.StreamHandler()

# Define a format string
formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s', datefmt='%Y-%m-%d %H:%M:%S')

# Attach formatter to handler, and handler to logger
handler.setFormatter(formatter)
logger.addHandler(handler)

logger.info("Server started on port 8080.")
```

## 4. Handlers
Handlers send the log records (created by loggers) to their appropriate destinations. A single logger can have multiple handlers.

StreamHandler: Sends logs to streams like sys.stdout or sys.stderr (the console).

FileHandler: Sends logs to a disk file.

SMTPHandler: Sends log emails to system administrators for critical errors.

HTTPHandler: Sends logs to a web server via GET or POST.

```python
import logging

logger = logging.getLogger("MultiHandlerApp")
logger.setLevel(logging.INFO)

# Handler 1: Log to console
console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)

# Handler 2: Log errors and above to a file
file_handler = logging.FileHandler("app_errors.log")
file_handler.setLevel(logging.ERROR)

# Add handlers to logger
logger.addHandler(console_handler)
logger.addHandler(file_handler)

logger.info("This goes to the console.")
logger.error("This goes to both the console and the log file!")
```

## 5. Rotating Logs
If an application runs continuously, a standard FileHandler will write to a single file forever, eventually consuming all disk space or creating a file too large to open. Python provides rotating handlers in the logging.handlers module to manage log file sizes automatically.

RotatingFileHandler: Rotates the log file based on maximum file size.

TimedRotatingFileHandler: Rotates the log file based on a specific time interval (e.g., daily, hourly).

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("RotatingApp")
logger.setLevel(logging.INFO)

# Rotate file when it reaches 2 Megabytes, keeping up to 3 backup files
handler = RotatingFileHandler("app.log", maxBytes=2*1024*1024, backupCount=3)
formatter = logging.Formatter('%(asctime)s - %(levelname)s - %(message)s')
handler.setFormatter(formatter)

logger.addHandler(handler)

logger.info("Logging with automatic rotation enabled.")
```
