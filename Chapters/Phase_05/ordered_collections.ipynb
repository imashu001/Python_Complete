# 1. Lists & Built-in Methods
A list is an ordered, mutable sequence of items. Lists can hold mixed data types, including numbers, strings, and even other lists.

## Common List Methods
Since lists are mutable, many of these methods modify the list in-place rather than creating a new one.

### Adding Elements:

    lst.append(item): Adds an item to the end of the list.

    lst.extend(iterable): Appends elements from another iterable to the end.

    lst.insert(index, item): Inserts an item at a specified index.

### Removing Elements:

    lst.remove(item): Removes the first occurrence of an item (raises ValueError if not found).

    lst.pop(index): Removes and returns the item at the given index (defaults to the last item).

    lst.clear(): Removes all items from the list.

### Ordering & Searching:

    lst.sort(): Sorts the list in-place (use sorted(lst) to return a new list).

    lst.reverse(): Reverses the list in-place.

    lst.index(item): Returns the index of the first occurrence.

    lst.count(item): Counts how many times an item appears.

```python
fruits = ["apple", "banana", "cherry"]

fruits.append("orange")  # ['apple', 'banana', 'cherry', 'orange']
fruits.insert(1, "blueberry")  # ['apple', 'blueberry', 'banana', 'cherry', 'orange']

popped_item = fruits.pop()  # Removes 'orange', returns it
fruits.sort()  # Alphabetizes the list in-place
```

## 2. Copying: Shallow vs. Deep Copies
Understanding how Python assigns and copies lists is critical to avoiding hard-to-find bugs.

Assignment (=): Does not create a copy. It simply creates a new reference pointing to the exact same list in memory. Changing one affects both.

Shallow Copy (copy(), list(), [:]): Creates a new container object, but the nested objects inside still point to the same memory references.

Deep Copy (copy.deepcopy()): Recursively copies everything, creating entirely independent duplicate objects down to the deepest nested level.

```python
import copy

original = [[1, 2], [3, 4]]

# Shallow copy using slicing
shallow = original[:]
shallow[0][0] = 99  # Modifies the nested list inside both!
print(original)  # Output: [[99, 2], [3, 4]]

# Deep copy
original_deep = [[1, 2], [3, 4]]
deep = copy.deepcopy(original_deep)
deep[0][0] = 99
print(original_deep)  # Output: [[1, 2], [3, 4]] (Original remains unaffected)
```

### 3. Nested Lists
Lists can contain other lists, allowing you to represent multi-dimensional data structures like matrices, grids, or game boards (e.g., Tic-Tac-Toe or chess grids).

To access elements in a nested list, you chain your index lookups (matrix[row][col]).

```python
# A 3x3 matrix (List of lists)
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

# Accessing row 1, column 2 (value 6)
print(matrix[1][2])  # Output: 6

# Iterating through a nested list
for row in matrix:
  for item in row:
    print(item, end=" ")
  print()
```

### 4. Tuples
A tuple is an ordered, immutable sequence. Once a tuple is created, you cannot add, remove, or modify its elements.

**Why Use Tuples?**
**Data Integrity:** Protects data that shouldn't change throughout the program's lifecycle.

**Performance:** Tuples are slightly more memory-efficient and faster to create than lists.

**Dictionary Keys:** Because they are immutable (and hashable), tuples can be used as keys in dictionaries (unlike lists).

```python
# Creating a tuple
coordinates = (10.0, 20.5)

# Single-element tuple requires a trailing comma!
not_a_tuple = (5)  # This is just an integer in parentheses
actual_tuple = (5,)  # This is a tuple

# Tuple Unpacking
x, y = coordinates
print(f"X: {x}, Y: {y}")
```
