# 1. Sets
A set is an unordered collection of unique elements. Because sets are unordered, they do not support indexing or slicing. However, they are optimized for membership testing and mathematical set operations.

## Common Set Operations
**Adding & Removing:** set.add(item), set.remove(item) (raises error if missing), set.discard(item) (safe removal, no error).

**Mathematical Operations:**

**Union (| or .union()):** Combines elements from both sets.

**Intersection (& or .intersection()): Keeps only elements found in both sets.

**Difference (- or .difference()):** Keeps elements in the first set but not the second.

**Symmetric Difference (^ or .symmetric_difference()):** Keeps elements in either set, but not both.

```python
# Creating sets and removing duplicates
numbers = [1, 2, 2, 3, 4, 4, 5]
unique_numbers = set(numbers)
print(unique_numbers)  # Output: {1, 2, 3, 4, 5}

set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print(set_a | set_b)  # Union: {1, 2, 3, 4, 5, 6}
print(set_a & set_b)  # Intersection: {3, 4}
print(set_a - set_b)  # Difference: {1, 2}
```

# 2. Dictionaries & Methods
A dictionary is a collection of key-value pairs. Keys must be unique and immutable, while values can be of any data type and duplicated.

## Essential Dictionary Methods
dict.get(key, default): Safely retrieves a value; returns default if the key doesn't exist instead of raising a KeyError.

dict.pop(key): Removes the key and returns its value.

dict.update(other_dict): Merges another dictionary or iterable of key-value pairs into the current one.

dict.setdefault(key, default): If the key exists, returns its value. If not, inserts the key with the given default value and returns it.

```python
student = {"name": "Alice", "age": 22, "major": "Computer Science"}

# Safe retrieval using .get()
gpa = student.get("gpa", 0.0)  # Returns 0.0 since 'gpa' doesn't exist

# Adding/Updating
student["gpa"] = 3.8
student.update({"age": 23, "graduated": False})
```

# 3. Dictionary Views
Methods like .keys(), .values(), and .items() return dictionary views rather than static lists.

**Dynamic Behavior:** Views reflect changes made to the underlying dictionary in real-time.

**Set-like Operations:** Dictionary keys and items support set operations (like intersections and unions) if the values are hashable.

```python
prices = {"apple": 1.20, "banana": 0.50, "cherry": 2.50}

keys_view = prices.keys()
print(keys_view)  # Output: dict_keys(['apple', 'banana', 'cherry'])

# Modify the dictionary
prices["orange"] = 0.80

# The view updates automatically!
print(keys_view)  # Output: dict_keys(['apple', 'banana', 'cherry', 'orange'])
```

# 4. Hashing & Mutability

Why can dictionaries have integers, strings, and tuples as keys, but not lists or sets? The answer lies in Hashing.

**What is a Hash?** 
A hash function takes an object and converts it into a fixed-size integer (a hash value).

Python uses these hash values to store and look up keys in dictionaries and sets with incredible efficiency ($O(1)$ constant time complexity).

**The Rule of Immutability**

For a hash value to remain valid, the object's contents **must never change** over its lifetime.

**Mutable objects** (like lists and dicts) can change their contents, meaning their hash value would change—breaking the dictionary lookup mechanism. Therefore, Python designates mutable objects as unhashable.

**Immutable objects** (like ints, floats, strings, and tuples containing only immutable items) are hashable and can safely be used as dictionary keys or set elements.

### **Pro Tip:**
If you ever need a composite key for a dictionary (like a 2D coordinate [x, y]), use a tuple (x, y) instead of a list [x, y], because tuples are hashable!
