## 1. Strings & Built-in Methods
Strings in Python are sequences of Unicode characters enclosed in single ('...'), double ("..."), or triple ("""...""" or '''...''') quotes. Strings are immutable, meaning once created, their characters cannot be changed in place.

### Common String Methods
Python provides a rich set of built-in methods to manipulate strings. These methods always return a new string and leave the original unchanged.

Case Modification:

    s.lower(), s.upper(), s.title(), s.capitalize()

Searching & Replacing:

    s.find(sub) (returns index or -1), s.index(sub) (returns index or raises ValueError)

    s.count(sub), s.replace(old, new)

Whitespace & Stripping:

    s.strip(), s.lstrip(), s.rstrip() (removes leading/trailing whitespace or specified characters)

Splitting & Joining:

    s.split(sep): Breaks a string into a list of substrings based on a delimiter.

    sep.join(iterable): Combines a collection of strings into a single string using sep as a glue.

```python
text = "  python programming is fun!  "
cleaned = text.strip()  # "python programming is fun!"
capitalized = cleaned.capitalize()  # "Python programming is fun!"
replaced = cleaned.replace("fun", "powerful")  # "python programming is powerful!"

words = cleaned.split(" ")  # ['python', 'programming', 'is', 'fun!']
joined = "-".join(words)  # "python-programming-is-fun!"
```

## 2. String Slicing
Slicing allows you to extract portions of a string using the syntax [start:stop:step].

    start: Inclusive starting index (default is 0).

    stop: Exclusive ending index (default is the end of the string).

    step: The stride or jump between characters (default is 1).

### Negative Indices & Reversing
Python supports negative indexing, where -1 represents the last character. A negative step reverses the direction of the slice.

```python
s = "PythonMaster"

print(s[0:6])  # "Python" (from index 0 up to, but not including, 6)
print(s[6:])  # "Master" (from index 6 to the end)
print(s[:6])  # "Python" (from start up to index 6)
print(s[::2])  # "PtoMstr" (every second character)
print(s[::-1])  # "retsaMnohtyP" (reverses the string entirely)
```

## 3. Formatting & f-strings
Formatting allows you to embed variables and expressions directly inside strings.

### Modern f-strings (Formatted String Literals)
Introduced in Python 3.6, f-strings are prefixed with f or F and use curly braces {} to evaluate expressions at runtime.

```python
name = "Alice"
score = 95.5

# Basic f-string
print(f"Hello, {name}! Your score is {score}.")

# Inline expressions and formatting modifiers
print(f"Next year, you will be {25 + 1} years old.")
print(f"Pi approximated: {3.14159265:.2f}")  # Formats to 2 decimal places: 3.14
print(f"Padded number: {42:05d}")  # Output: 00042
```

### Older Formatting Methods (For context)
    % formatting: print("Hello, %s!" % name) (Older style, borrowed from C).

    .format() method: print("Hello, {}!".format(name)) (Pre-f-string standard, still useful for template storage).

## 4. Encoding & Decoding
Computers do not understand characters; they understand numbers (bytes). Encoding is the process of converting a human-readable string into a sequence of bytes, while Decoding does the reverse.

    Python strings use Unicode (capable of representing nearly every language and emoji).

    When writing to a file, sending data over a network, or interacting with low-level APIs, text must be encoded into a standard byte representation, most commonly UTF-8.

```python
# A standard Python string (Unicode)
original_text = "Hello, 🌍!"

# Encoding: converting string to bytes using UTF-8
encoded_bytes = original_text.encode("utf-8")
print(encoded_bytes)
# Output: b'Hello, \xf0\x9f\x8c\x8d!' (Notice the 'b' prefix indicating bytes)

# Decoding: converting bytes back to a string
decoded_text = encoded_bytes.decode("utf-8")
print(decoded_text)  # Output: Hello, 🌍!
```

### Common Pitfall: Trying to concatenate a str and a bytes object directly will result in a TypeError. Always ensure data is decoded before text processing or encoded before binary operations.
