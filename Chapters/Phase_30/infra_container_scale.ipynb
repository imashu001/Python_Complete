## 1. Docker (Containerization)
Docker packages an application and all its dependencies (Python runtime, OS libraries, third-party packages) into a single isolated container, ensuring it runs identically on your local machine, a testing server, or a cloud cluster.

```python
# Example Dockerfile for a FastAPI Python application
FROM python:3.11-slim

WORKDIR /app

# Install dependencies using uv for maximum speed
COPY pyproject.toml uv.lock ./
RUN pip install uv && uv sync --frozen

COPY . .

EXPOSE 8000
CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]
```

## 2. Kubernetes (K8s Basics)
When managing dozens or hundreds of Docker containers across multiple servers, manual management becomes impossible. Kubernetes is an open-source container orchestration platform that automates deployment, scaling, and container operations.

Pod: The smallest deployable unit in Kubernetes, typically running one or more tightly coupled containers.

Deployment: Manages the rollout and scaling of Pods, ensuring a specified number of replicas are running at all times.

Service: An abstraction that defines a logical set of Pods and enables a stable networking endpoint to access them.

## 3. CI/CD (Continuous Integration & Continuous Deployment)
CI/CD automates the software delivery lifecycle so that code changes are automatically tested, built, and deployed to production.

Continuous Integration (CI): Every time code is pushed to version control, automated pipelines run linters (ruff), security scans, and test suites (pytest).

Continuous Deployment (CD): Once tests pass successfully, the pipeline builds a Docker image, pushes it to a container registry, and updates the production environment automatically.

## 4. Scaling & Load Balancing
As user traffic grows, infrastructure must scale to prevent downtime and maintain low latency.

Vertical Scaling (Scaling Up): Adding more CPU, RAM, or storage to a single server instance. It is simple to implement but has hard physical and financial limits.

Horizontal Scaling (Scaling Out): Adding more server instances (or Kubernetes pods) behind a shared pool. This provides near-infinite scaling capabilities and fault tolerance.

Load Balancing: A load balancer (e.g., Nginx, AWS ALB, HAProxy) sits in front of your application servers, distributing incoming user traffic across multiple backend nodes using algorithms like Round Robin or Least Connections.
