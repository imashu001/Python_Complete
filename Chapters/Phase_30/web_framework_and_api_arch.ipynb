# Web Frameworks & API Architectures
Building production-grade applications requires a firm grasp of web architecture and the tools used to expose business logic. In this first part of Phase 30, we cover REST API design principles alongside Python's three primary web frameworks: FastAPI, Flask, and Django.

## 1. REST APIs (Representational State Transfer)
REST is an architectural style for designing networked applications using standard HTTP protocols.

**Statelessness:** Every request from a client must contain all the information necessary for the server to understand and fulfill it. Servers do not store client session data between requests.

**Standard HTTP Methods:**

GET: Retrieve a resource (safe and idempotent).

POST: Create a new resource.

PUT / PATCH: Update an existing resource (complete replacement vs. partial update).

DELETE: Remove a resource.

**Resource-Centric URIs:** Endpoints should use nouns rather than verbs (e.g., /api/v1/users/{id} instead of /api/v1/getUser).

## 2. FastAPI
FastAPI is a modern, high-performance web framework built for building APIs with Python, leveraging standard Python type hints.

**Key Strengths:** Automatic interactive documentation (/docs via Swagger UI and /redoc), blazing-fast execution speeds (on par with Node.js and Go via Starlette and Pydantic), and native support for asynchronous programming (async/def).

```python
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel

app = FastAPI()

class Item(BaseModel):
    name: str
    price: float
    is_offer: bool = None

@app.post("/items/")
def create_item(item: Item):
    # Pydantic automatically validates request body data
    return {"message": "Item created successfully", "item": item}

@app.get("/items/{item_id}")
def read_item(item_id: int, q: str = None):
    return {"item_id": item_id, "query": q}
```

## 3. Flask
Flask is a lightweight "micro-framework" written in Python. It gives developers total freedom and flexibility to choose their own database libraries, authentication layers, and project structures.

**Key Strengths:** Minimalistic, highly extensible, easy to learn, and perfect for small services, prototypes, or custom micro-applications.

```python
from flask import Flask, jsonify, request

app = Flask(__name__)

@app.route("/api/v1/health", methods=["GET"])
def health_check():
    return jsonify({"status": "healthy", "service": "running"}), 200

if __name__ == "__main__":
    app.run(debug=True, port=5000)
```

## 4. Django (Basics)
Django is a high-level, "batteries-included" Python web framework designed to encourage rapid development and clean, pragmatic design.

**Key Strengths:** Comes pre-packed with an Object-Relational Mapper (ORM), an admin dashboard, user authentication, security protection (CSRF, SQL injection, XSS defense), and a robust migration system.

**Architecture:** Follows the MVT (Model-View-Template) architectural pattern.

```python
# Django models example (models.py)
from django.db import models

class Article(models.Model):
    title = models.CharField(max_length=200)
    content = models.TextField()
    created_at = models.DateTimeField(auto_now_add=True)

    def __str__(self):
        return self.title
```

### Summary Comparison: When to Use Which?
Use FastAPI for high-throughput microservices, real-time APIs, machine learning model serving, and automatic data validation.

Use Flask for lightweight scripts, simple webhooks, or micro-services where you want absolute control over your stack.

Use Django for data-heavy monoliths, content management systems, applications requiring a built-in admin panel, and rapid MVP launches.
