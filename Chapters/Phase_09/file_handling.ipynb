## 1. Reading & Writing Files & The with Statement
To interact with files on your hard drive, Python provides the built-in open() function.

### File Modes
'r': Read (default mode; raises an error if the file doesn't exist).

'w': Write (creates a new file or truncates/overwrites an existing one).

'a': Append (adds data to the end of an existing file without deleting old content).

'b': Binary mode (used with images, audio, etc., e.g., 'rb' or 'wb').

### The with Statement (Context Manager)
Traditionally, when you opened a file, you had to remember to close it manually (file.close()) to prevent memory leaks or locked files. The with statement acts as a context manager that automatically closes the file as soon as the indented block of code finishes executing, even if an error occurs.

```python
# Writing to a file safely using 'with'
with open("notes.txt", "w") as file:
  file.write("Hello, Python file handling!\n")
  file.write("Learning file operations is fun.")

# Reading from a file
with open("notes.txt", "r") as file:
  content = file.read()  # Reads the entire file as a string
  print(content)

# Reading line by line (memory efficient for large files)
with open("notes.txt", "r") as file:
  for line in file:
    print(line.strip())
```

## 2. CSV Files
Comma-Separated Values (CSV) files are widely used for tabular data (like spreadsheets or database exports). Python provides a built-in module called csv to handle them easily.

```python
import csv

# Writing to a CSV file
data = [
    ["Name", "Age", "Role"],
    ["Alice", 28, "Developer"],
    ["Bob", 34, "Designer"],
]

with open("people.csv", "w", newline="") as file:
  writer = csv.writer(file)
  writer.writerows(data)

# Reading from a CSV file using dictionaries for easy column access
with open("people.csv", "r") as file:
  reader = csv.DictReader(file)
  for row in reader:
    print(f"{row['Name']} is a {row['Role']} aged {row['Age']}.")
```

## 3. JSON Files
JSON (JavaScript Object Notation) is the universal format for exchanging structured data over the web and APIs. Python's built-in json module maps JSON objects directly to Python dictionaries and lists.

json.dump() / json.dumps(): Serializes Python objects into JSON strings/files.

json.load() / json.loads(): Deserializes JSON strings/files back into Python objects.

```python
import json

student = {
    "name": "Charlie",
    "grades": [85, 90, 95],
    "active": True,
    "address": {"city": "Seattle", "zip": "98101"},
}

# Writing (dumping) Python dict to a JSON file
with open("student.json", "w") as file:
  json.dump(student, file, indent=4)  # indent makes it human-readable

# Reading (loading) JSON file back into a Python dictionary
with open("student.json", "r") as file:
  loaded_data = json.load(file)
  print(loaded_data["name"])  # Output: Charlie
  print(loaded_data["grades"])  # Output: [85, 90, 95]
```

## 4. Pickle
While JSON is great for human-readable, cross-language data, Pickle is Python's built-in module for binary serialization.

**What it does:** It can serialize almost any complex Python object (including custom class instances, data frames, or trained machine learning models) directly into a byte stream.

**Security Warning:** Never unpickle data received from an untrusted or unauthenticated source, as it can execute arbitrary malicious code.

```python
import pickle

complex_data = {
    "numbers": [1, 2, 3],
    "custom_object": None,  # Can hold complex types
}

# Writing binary pickle file
with open("data.pkl", "wb") as file:
  pickle.dump(complex_data, file)

# Reading binary pickle file
with open("data.pkl", "rb") as file:
  restored_data = pickle.load(file)
  print(restored_data)
```

## 5. YAML
YAML ("YAML Ain't Markup Language") is a human-friendly data serialization standard frequently used for configuration files (like Docker compose, CI/CD pipelines, or settings files).

### Note: YAML is not part of Python's standard library, so you need to install the external PyYAML package via terminal (pip install PyYAML) before using it.

```python
# Run in terminal first: pip install PyYAML
import yaml

config = {
    "server": {"host": "localhost", "port": 8080},
    "debug_mode": True,
    "allowed_users": ["admin", "developer"],
}

# Writing to a YAML file
with open("config.yaml", "w") as file:
  yaml.dump(config, file, default_flow_style=False)

# Reading from a YAML file
with open("config.yaml", "r") as file:
  loaded_config = yaml.safe_load(file)
  print(loaded_config["server"]["port"])  # Output: 8080
```
