# Phase 2: Operators

Operators are special symbols or keywords in Python used to perform operations on variables and values. 

---

## 1. Arithmetic Operators

Used to perform standard mathematical computations.

| Operator | Name | Description | Example (`x = 10, y = 3`) | Result |
| :--- | :--- | :--- | :--- | :--- |
| `+` | Addition | Adds values together | `x + y` | `13` |
| `-` | Subtraction | Subtracts right operand from left | `x - y` | `7` |
| `*` | Multiplication | Multiplies values | `x * y` | `30` |
| `/` | Division | Divides left operand by right (returns float) | `x / y` | `3.3333...` |
| `//` | Floor Division | Divides and rounds down to the nearest integer | `x // y` | `3` |
| `%` | Modulus | Returns the remainder of division | `x % y` | `1` |
| `**` | Exponentiation | Raises left operand to the power of right | `x ** y` | `1000` |

```python
# Arithmetic examples
print(10 / 3)   # Output: 3.3333333333333335 (Float division)
print(10 // 3)  # Output: 3 (Floor division)
print(2 ** 4)   # Output: 16 (2 to the power of 4)

## 2. Comparison (Relational) Operators
Used to compare two values. They always evaluate to a boolean value (True or False).

| Operator | Name | "Example (x = 5, y = 10)" | Result
| :--- | :--- | :--- | :--- |
| == | Equal to |x == y | False
| != |Not equal to | x != y | True
| > |Greater than |x > y | False
| < | Less than | x < y | True
| >= | Greater than or equal to | x >= 5 | True
| <= | Less than or equal to |y <= 10 | True

## 3. Assignment Operators
Used to assign values to variables, often combined with an arithmetic operation.

| Operator | Equivalent Expression | Example
| :--- | :--- | :--- |
| = | x = 5 | x = 5
| += | x = x + 3 | x += 3 (x becomes 8)
| -= | x = x - 2 | x -= 2
| *= | x = x * 4 | x *= 4
| /= | x = x / 2 | x /= 2
| //= | x = x // 3 | x //= 3
| %= | x = x % 2 | x %= 2
| **= | x = x ** 2 | x **= 2

````python
score = 10
score += 5  # Equivalent to: score = score + 5
print(score)  # Output: 15
````

## 4. Logical Operators
Used to combine conditional statements and evaluate truth values.

| Operator | Description | Example
| :--- | :--- | :--- |
| and | Returns True if both statements are true | (5 > 3) and (10 > 5) → True
| or | Returns True if at least one statement is true | (5 > 3) or (10 < 5) → True
| not | "Reverses the result (turns True to False, and vice versa)" | not(5 > 3) → False

Short-Circuit Evaluation: Python evaluates logical operators from left to right. For and, if the first condition is False, Python stops evaluating and immediately returns False. For or, if the first condition is True, it stops and returns True.

## 5. Bitwise Operators
Used to compare binary numbers (bits 0 and 1).

| Operator | Name | Description
| :--- | :--- | :--- |
| & | AND | Sets each bit to 1 if both bits are 1
| \| | OR | Sets each bit to 1 if one of two bits is 1
| ^ | XOR | Sets each bit to 1 if only one of two bits is 1
| ~ | NOT | Inverts all the bits
| << | Zero fill left shift | Shift left by pushing zeros in from the right
| >> | Signed right shift | Shift right by pushing copies of the leftmost bit in from the left

````python
# Example: Binary operations on 5 (0101) and 3 (0011)
print(5 & 3)  # Output: 1 (Binary: 0001)
print(5 | 3)  # Output: 7 (Binary: 0111)
````

## 6. Membership Operators
Used to test if a sequence (such as a string, list, tuple, set, or dictionary) contains a specific value.

| Operator | Description | Example
| :--- | :--- | :--- |
| in | Returns True if value is found in the sequence | """a"" in ""apple"" → True"
| not in | Returns True if value is not found in the sequence | """x"" not in ""apple"" → True"

````python
colors = ["red", "green", "blue"]
print("green" in colors)  # Output: True
print("yellow" not in colors)  # Output: True
````

## 7. Identity Operators (is vs ==)
It is vital to understand the difference between identity and equality in Python:

== (Equality): Checks if the values of two objects are equal.

is (Identity): Checks if two variables point to the exact same object in memory (same memory address / id()).

````python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

# Values are identical
print(a == b)  # Output: True (contents are identical)

# Memory references are different
print(a is b)  # Output: False (different objects in memory)

# c points to the exact same object as a
print(a is c)  # Output: True
````

## 8. Operator Precedence
When an expression contains multiple operators, Python evaluates them based on a strict order of precedence (from highest to lowest). Operations with higher precedence are evaluated first.

Order of Precedence (High to Low):
Parentheses (): Evaluated first (use them to force order).

Exponentiation **

Unary plus, minus, bitwise NOT: +x, -x, ~x

Multiplication, Division, Floor Division, Modulus: *, /, //, %

Addition and Subtraction: +, -

Bitwise shifts: <<, >>

Bitwise AND: &

Bitwise XOR: ^

Bitwise OR: \|

Comparisons, Membership, Identity: ==, !=, >, <, >=, <=, is, is not, in, not in

Logical NOT: not

Logical AND: and

Logical OR: or

````python
# Without parentheses (multiplication happens before addition)
result1 = 10 + 2 * 3  # Output: 16 (10 + 6)

# With parentheses (forces addition to happen first)
result2 = (10 + 2) * 3  # Output: 36 (12 * 3)
````
