# Linear Data Structures & Interview Patterns

Linear data structures organize elements sequentially, where each element is connected to its previous and next neighbor. In technical interviews, mastering these fundamental structures—and knowing how Python implements them under the hood—is non-negotiable.

## 1. Arrays & Strings

In Python, the closest built-in equivalent to a traditional contiguous array is the List (list). Lists are dynamic arrays, meaning they automatically resize themselves when they run out of preallocated memory.

### Core Python Considerations & Time Complexities:

Lookup / Access (arr[i]): $O(1)

$Append / Pop from End: $O(1)$ amortized

Insert / Delete from Beginning or Middle: $O(N)$ (because all subsequent elements must be shifted in memory)

### Essential Interview Patterns:

Two Pointers: Moving two indices toward each other (or in the same direction) to solve problems in $O(N)$ time instead of $O(N^2)$ nested loops.

Sliding Window: Maintaining a dynamic window (subarray or substring) that expands or shrinks to solve range-based problems.

```python
# Example: Two Pointers - Check if a string is a palindrome
def is_palindrome(s: str) -> bool:
    # Clean string: lowercase and keep only alphanumeric characters
    cleaned = "".join(ch.lower() for ch in s if ch.isalnum())
    
    left, right = 0, len(cleaned) - 1
    while left < right:
        if cleaned[left] != cleaned[right]:
            return False
        left += 1
        right -= 1
    return True
```

## 2. Linked Lists
A Linked List consists of nodes where each node contains a value and a reference (pointer) to the next node. Unlike arrays, linked lists do not require contiguous memory blocks, allowing for $O(1)$ insertions and deletions at the head, but $O(N)$ search time.

### Python Implementation:

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

# Reversing a Linked List (Classic Interview Question)
def reverse_linked_list(head: ListNode) -> ListNode:
    prev = None
    curr = head
    
    while curr:
        next_temp = curr.next  # Temporarily store next node
        curr.next = prev       # Reverse the pointer
        prev = curr            # Move prev forward
        curr = next_temp       # Move curr forward
        
    return prev  # New head of the reversed list
```

### Essential Interview Patterns:
Fast and Slow Pointers (Tortoise and Hare): Two pointers moving at different speeds to find cycles in a linked list or locate the middle element.

Dummy Head Node: Creating a dummy node before the actual head simplifies edge cases when inserting or deleting nodes.

## 3. Stack

A Stack is a LIFO (Last-In, First-Out) data structure. Think of a stack of plates: the last plate you put on top is the first one you take off.

### Python Implementation:
In Python, you should never use a standard list with insert(0, ...) for stacks where performance matters, but standard .append() and .pop() operate in $O(1)$ time at the end of the list. Alternatively, use collections.deque for optimal performance.

```python
from collections import deque

class Stack:
    def __init__(self):
        self.items = deque()

    def push(self, item):
        self.items.append(item)

    def pop(self):
        return self.items.pop() if self.items else None

    def peek(self):
        return self.items[-1] if self.items else None

# Classic Interview Problem: Valid Parentheses
def isValid(s: str) -> bool:
    stack = []
    mapping = {")": "(", "}": "{", "]": "["}
    
    for char in s:
        if char in mapping:
            top_element = stack.pop() if stack else '#'
            if mapping[char] != top_element:
                return False
        else:
            stack.append(char)
            
    return not stack
```

### Essential Interview Patterns:

**Monotonic Stack:** A stack that maintains its elements in a strictly increasing or decreasing order. Used heavily for finding the "Next Greater Element" in an array.

## 4. Queue
A Queue is a FIFO (First-In, First-Out) data structure. Think of a line of people waiting for a ticket: the first person in line is the first one served.

### Python Implementation:
Again, avoid using standard lists as queues because popping from the front (pop(0)) forces an $O(N)$ memory shift. Always use collections.deque for true $O(1)$ operations from both ends.

```python
from collections import deque

class Queue:
    def __init__(self):
        self.items = deque()

    enqueue = deque.append

    def dequeue(self):
        return self.items.popleft() if self.items else None

    def is_empty(self):
        return not self.items
```

### Essential Interview Patterns:
**Breadth-First Search (BFS):** Queues are the core engine behind BFS algorithms used to traverse trees and graphs level by level.

### Best Practices & Common Pitfalls
**Choose the Right Python Container:** Using list.pop(0) for a queue turns an $O(1)$ operation into an $O(N)$ bottleneck. Always remember collections.deque for queues and double-ended operations.

**Watch Out for None Pointers:** When manipulating linked lists, always verify that curr and curr.next are not None before attempting to access their attributes, or you will trigger an AttributeError.
