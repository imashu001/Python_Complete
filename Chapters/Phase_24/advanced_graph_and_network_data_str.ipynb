## 1. Graphs
A graph is a collection of nodes (vertices) connected by links (edges). Graphs can be directed (edges have a single direction, like Twitter followers) or undirected (edges go both ways, like Facebook friends), and they can be weighted or unweighted.

### Graph Representations in Python:
Graphs are typically represented in one of two ways:

**Adjacency List:** A dictionary where each node maps to a list of its neighbors. This is the most common and memory-efficient representation in Python interviews.

**Adjacency Matrix:** A 2D grid (matrix[i][j]) representing edge weights between nodes.

```python
# Adjacency List Representation
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}
```

### Essential Interview Patterns & Traversals:
**Breadth-First Search (BFS):** Explores neighbors level by level using a queue (deque). Ideal for finding the shortest path in unweighted graphs.

**Depth-First Search (DFS):** Explores as deep as possible along each branch before backtracking, using recursion or an explicit stack. Ideal for detecting cycles, connected components, or path existence.

```python
from collections import deque

# BFS Implementation for Shortest Path / Traversal
def bfs(graph, start_node):
    visited = set()
    queue = deque([start_node])
    visited.add(start_node)
    
    result = []
    while queue:
        node = queue.popleft()
        result.append(node)
        
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
                
    return result
```

### Advanced Algorithms to Know:
**Dijkstra’s Algorithm:** Uses a priority queue (heapq) to find the shortest path in a weighted graph with non-negative edge weights.

**Topological Sort:** Orders the nodes in a Directed Acyclic Graph (DAG) such that for every directed edge $u \to v$, node $u$ comes before $v$ (crucial for course scheduling / dependency problems).

## 2. Union-Find (Disjoint Set Union - DSU)

Union-Find is a data structure that tracks a set of elements partitioned into a number of disjoint (non-overlapping) subsets. It supports two primary operations in nearly constant time ($O(1)$ amortized):

**Find:** Determine which subset a particular element is in (often used to check if two elements share the same root/component).

**Union:** Join two subsets into a single subset.

### Python Implementation:

DSU relies heavily on two optimizations: Path Compression (flattening the tree structure during find) and Union by Rank/Size (attaching smaller trees under larger trees).

```python
class UnionFind:
    def __init__(self, n: int):
        self.parent = list(range(n))
        self.rank = [1] * n

    def find(self, i: int) -> int:
        # Path compression: point node directly to the root
        if self.parent[i] == i:
            return i
        self.parent[i] = self.find(self.parent[i])
        return self.parent[i]

    def union(self, i: int, j: int) -> bool:
        root_i = self.find(i)
        root_j = self.find(j)
        
        if root_i != root_j:
            # Union by rank
            if self.rank[root_i] > self.rank[root_j]:
                self.parent[root_j] = root_i
            elif self.rank[root_i] < self.rank[root_j]:
                self.parent[root_i] = root_j
            else:
                self.parent[root_j] = root_i
                self.rank[root_i] += 1
            return True  # Successfully merged
        return False     # Already in the same set (Cycle detected!)
```

### Essential Interview Patterns:
Cycle Detection in Undirected Graphs: If an edge connects two nodes that already share the same root in DSU, adding that edge creates a cycle.

Kruskal's Algorithm: Uses Union-Find to find the Minimum Spanning Tree (MST) of a graph by sorting all edges by weight and greedily adding them if they don't form a cycle.

Connected Components: Counting how many isolated subnetworks exist in a graph.

### Best Practices & Common Pitfalls

Always Track Visited Nodes: When traversing graphs with cycles (like standard cyclic graphs), failing to maintain a visited set will lead to infinite loops and stack overflow errors.

Watch Out for Disconnected Graphs: Real-world graphs are not always fully connected. When running traversals like BFS or DFS, ensure you loop through all nodes in the graph to catch isolated components that weren't reached from your starting node.

Optimize DSU Lookups: Never implement a raw Union-Find without path compression; without it, the tree degenerates into a linear linked list, ruining the $O(1)$ performance guarantee.
