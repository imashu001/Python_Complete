# Hash-Based & Hierarchical Structures
In this section, we move beyond linear sequences into hash-based storage for lightning-fast lookups and hierarchical tree structures used to model parent-child relationships and prioritized data.

## 1. Hash Tables (Dictionaries & Sets in Python)
A Hash Table stores key-value pairs and computes an index (hash) using a hash function to map keys to slots. This allows for near-instantaneous data retrieval.

### Core Python Considerations & Time Complexities:
Average Case Lookup, Insertion, Deletion: $O(1)$
Worst Case (Hash Collisions): $O(N)$

```python
# Classic Interview Pattern: Two Sum using a Hash Map
def two_sum(nums: list[int], target: int) -> list[int]:
    seen = {}  # Value -> Index mapping
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    return []
```

### Essential Interview Patterns:
**Frequency Counting:** Using hash maps or collections.Counter to track frequencies for anagrams, duplicates, or majority voting algorithms.

**Prefix/Running Sum Hash Maps:** Storing cumulative sums to find subarrays that sum to a specific target in $O(N)$ time.

## 2. Heap (Priority Queue)
A Heap is a specialized tree-based data structure that satisfies the heap property: in a Min-Heap, the parent is less than or equal to its children (so the root is always the minimum); in a Max-Heap, the parent is greater.Python Implementation:Python's built-in heapq module implements a Min-Heap using standard lists.

```python
import heapq

# Classic Interview Pattern: Find the Kth Largest Element in an Array
def find_kth_largest(nums: list[int], k: int) -> int:
    # Maintain a min-heap of size k
    min_heap = []
    for num in nums:
        heapq.heappush(min_heap, num)
        if len(min_heap) > k:
            heapq.heappop(min_heap)
            
    return min_heap[0]  # The root is the kth largest element
```

### Essential Interview Patterns:
Top K Elements: Using a heap of fixed size $K$ to find the smallest or largest elements in $O(N \log K)$ time instead of sorting the entire array ($O(N \log N)$).

Merge K Sorted Lists: Combining multiple sorted streams efficiently.

## 3. Trees & Binary Search Trees (BST)
A tree is a hierarchical structure consisting of nodes connected by edges, with a single root node at the top. A Binary Search Tree (BST) is a binary tree where every node satisfies the property: all values in the left subtree are less than the node's value, and all values in the right subtree are greater.

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

# Classic BST Validation
def isValidBST(root: TreeNode, low=float('-inf'), high=float('inf')) -> bool:
    if not root:
        return True
    if not (low < root.val < high):
        return False
    return (isValidBST(root.left, low, root.val) and 
            isValidBST(root.right, root.val, high))
```

### Essential Interview Patterns:
Tree Traversals:

BFS (Level Order): Using a queue (deque) to visit nodes level by level.

DFS (Pre-order, In-order, Post-order): Using recursion or an explicit stack. Note: An In-order traversal of a valid BST always yields sorted elements.

Recursion & Divide and Conquer: Breaking tree problems down into sub-problems on left and right subtrees.

## 4. Trie (Prefix Tree)
A Trie is a specialized tree used to store strings efficiently, particularly for retrieval by prefix. Every node represents a character, enabling ultra-fast autocomplete and spell-checking algorithms.

### Python Implementation:

```python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_end_of_word = False

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word: str) -> None:
        curr = self.root
        for char in word:
            if char not in curr.children:
                curr.children[char] = TrieNode()
            curr = curr.children[char]
        curr.is_end_of_word = True

    def search(self, word: str) -> bool:
        curr = self.root
        for char in word:
            if char not in curr.children:
                return False
            curr = curr.children[char]
        return curr.is_end_of_word
```

### Essential Interview Patterns:
Prefix Matching: Finding all words starting with a specific substring in $O(M)$ time (where $M$ is the length of the word), independent of the number of words stored in the Trie.

### Best Practices & Common Pitfalls
**Remember Python's Min-Heap limitation:** heapq only provides a min-heap. If you need a max-heap, you must invert your values (multiply numbers by -1) when pushing and popping.

**Beware of Unbalanced BSTs:** If a BST becomes completely skewed (resembling a linked list), operations degrade from $O(\log N)$ to $O(N)$. Interview questions will often test your awareness of self-balancing trees like AVL or Red-Black trees conceptually.
