# Phase 8: Modules & Packages
Welcome to Phase 8! As your Python programs grow beyond a single script, you need a way to organize, reuse, and structure your code across multiple files. In this chapter, you will learn how Python handles code modularity through modules, packages, and execution scopes.

## 1. Modules, Imports, and Aliases
A module is simply a single Python file (ending in .py) containing code—functions, classes, and variables—that you want to reuse in other files.

### Import Syntaxes
import module_name: Imports the entire module. You must access contents using dot notation (module_name.function()).

from module_name import item: Imports a specific function or variable directly into your current namespace.

from module_name import *: Imports everything (generally discouraged as it clutters your namespace and causes naming conflicts).

### Aliasing (as)
You can rename modules or functions upon import using the as keyword to avoid naming collisions or shorten long names.

```python
# Importing an entire module
import math

print(math.sqrt(16))  # Output: 4.0

# Importing a specific item with an alias
from math import factorial as fact

print(fact(5))  # Output: 120

# Importing and aliasing a module
import numpy as np

# (Assuming numpy is installed, you'd use np.array())
```

## 2. Package Structure & __init__.py
A package is a collection of modules organized in a directory hierarchy. To tell Python that a directory should be treated as a package, it contains a special initialization file: __init__.py.

Standard Package Layout

```python
my_project/
│
├── main.py
└── my_package/
    ├── __init__.py
    ├── math_tools.py
    └── string_tools.py
```

### The Role of __init__.py
**Python 3.3+ (Implicit Namespace Packages):** Technically, __init__.py is no longer strictly required for Python to treat a folder as a package. However, it is still heavily used and best practice.

**Initialization Code:** Code placed inside __init__.py runs automatically when the package is imported.

**Controlling Exports (__all__):** You can use __init__.py to define which modules or functions are exposed when someone writes from my_package import *.

```python
# Inside my_package/__init__.py
from .math_tools import add_numbers
from .string_tools import capitalize_text

__all__ = ["add_numbers", "capitalize_text"]
```

## 3. __name__ and __main__
Every Python module has a built-in variable called __name__.

When a file is run directly as the main program, Python sets __name__ to "__main__".

When a file is imported into another script as a module, Python sets __name__ to the actual filename (e.g., "math_tools").

### The Boilerplate Pattern
This mechanism allows you to write scripts that can both be imported safely into other files and executed directly as standalone programs for testing or command-line execution.

```python
# math_tools.py


def add(a, b):
  return a + b


def subtract(a, b):
  return a - b


# This block only runs if math_tools.py is executed directly
if __name__ == "__main__":
  print("Running tests for math_tools...")
  print(f"2 + 3 = {add(2, 3)}")
  print(f"5 - 2 = {subtract(5, 2)}")
```
