# Typing
Python is dynamically typed by nature, meaning variables can change types at runtime. However, as codebases grow, large applications benefit immensely from static type checking. Introduced in Python 3.5 and expanded heavily in recent versions, Python's typing system allows you to document, lint, and catch type-related bugs before your code ever runs (using tools like mypy).

## 1. Type Hints
Type hints allow you to explicitly declare the expected data types for function arguments, return values, and variables. They do not enforce types at runtime by themselves, but they improve code readability and enable powerful IDE auto-completion and static analysis.

```python
# Function with type hints for parameters and return value
def greet(name: str, age: int) -> str:
    return f"Hello {name}, you are {age} years old."

# Variable type annotations
score: float = 95.5
is_active: bool = True
```

## 2. Optional, Union, and Any
When an argument or variable can accept multiple types or might be missing entirely, Python provides special type constructs from the typing module (or built-in structures in Python 3.10+).

Union: Specifies that a value can be one of several types (in Python 3.10+, you can use the pipe operator | instead, e.g., int | str).

Optional: Indicates that a value can be of a specific type or None. Optional[str] is shorthand for Union[str, None] (or str | None).

Any: Opts out of type checking entirely for that specific variable or parameter, signaling that any type is acceptable.

```python
from typing import Optional, Union, Any

# Using Union (or int | float in Python 3.10+)
def process_number(num: Union[int, float]) -> float:
    return float(num)

# Using Optional (or str | None)
def get_middle_name(user_id: int) -> Optional[str]:
    # Returns a string name or None if not found
    return None
```

## 3. Generic Types
Generics allow you to write reusable classes and functions that work with multiple data types while preserving type safety for the items inside containers (like lists, dictionaries, or custom data structures).

```python
from typing import TypeVar, List

# Define a Type Variable
T = TypeVar('T')

def get_first_item(items: List[T]) -> T:
    return items[0]

# Mypy knows that if you pass a list of ints, the return type is an int
first_int = get_first_item([1, 2, 3])
```

## 4. Protocols (Structural Subtyping / Duck Typing)
Protocols allow you to define types based on behavior (what methods or attributes an object has) rather than explicit inheritance. This is Python's way of formalizing duck typing.

```python
from typing import Protocol

# Define a protocol for anything that can "render"
class Renderable(Protocol):
    def render(self) -> str:
        ...

def display_item(item: Renderable) -> None:
    print(item.render())

# Even though Book doesn't inherit from Renderable, it matches the protocol implicitly
class Book:
    def render(self) -> str:
        return "Rendering Book HTML..."

display_item(Book())  # Valid type checking!
```

## 5. Dataclasses
Introduced in Python 3.7, the @dataclass decorator automatically generates boilerplate methods (like \_\_init\_\_, \_\_repr\_\_, and \_\_eq\_\_) for classes that primarily store data.

```python
from dataclasses import dataclass, field
from typing import List

@dataclass
def Product:
    name: str
    price: float
    tags: List[str] = field(default_factory=list) # Safe default for mutable types

item = Product(name="Laptop", price=999.99, tags=["electronics", "computers"])
print(item)  # Output: Product(name='Laptop', price=999.99, tags=['electronics', 'computers'])
```

## 6. TypedDict
When working with dictionaries that have a fixed set of keys with specific, heterogeneous value types (such as parsing JSON payloads or API responses), TypedDict provides strong type checking for dictionary structures.

```python
from typing import TypedDict

class UserDict(TypedDict):
    id: int
    username: str
    email: Optional[str]

# Type checker will validate keys and value types
user: UserDict = {"id": 1, "username": "alice_dev", "email": "alice@example.com"}
```

## 7. Enums (Enumerations)
Enums are used to create a symbolic name for a set of constant values. They prevent magic strings or numbers from floating around your code, making states explicit and clean.

```python
from enum import Enum, auto

class Status(Enum):
    PENDING = auto()
    RUNNING = auto()
    COMPLETED = auto()
    FAILED = auto()

current_status = Status.RUNNING

if current_status == Status.RUNNING:
    print("Process is active...")
```

### Best Practices & Common Pitfalls
**Type Hints Aren't Runtime Enforcers:** Python does not throw a TypeError if you pass a string into a function annotated with int. If you need strict runtime validation, look into libraries like Pydantic.

**Use Modern Syntax When Possible:** Python 3.10+ introduced built-in union operators (int | str instead of Union[int, str]) and built-in collection generics (list[int] instead of typing.List[int]), which make code cleaner and reduce imports.

**Avoid Any Unless Necessary:** Using Any disables type checking for that object. Overusing it defeats the purpose of adding type hints in the first place.
