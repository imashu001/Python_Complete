Dunder methods (short for Double Underscore methods, e.g., __init__) are also commonly known as magic methods. They are special hooks pre-defined by Python that allow your custom classes to integrate seamlessly with built-in functions (len(), str(), hash()), operators (+, <, ==), and structural patterns (context managers, iterators). 


# 1. Representation & Initialization (__init__, __str__, __repr__)

These methods control how an object is created, initialized, and displayed to users or developers.__init__(self, ...): The initializer called after an object is created.  __str__(self): The informal, user-friendly string representation triggered by print() or str().  __repr__(self): The official, developer-friendly string representation meant for debugging and logging.  

```python
class Book:

  def __init__(self, title, author):
    self.title = title
    self.author = author

  def __str__(self):
    return f"'{self.title}' by {self.author}"

  def __repr__(self):
    return f"Book(title='{self.title}', author='{self.author}')"


b = Book("1984", "George Orwell")
print(str(b))  # Output: '1984' by George Orwell (calls __str__)
print(repr(b))  # Output: Book(title='1984', author='George Orwell') (calls __repr__)
```

# 2. Container & Length Protocol (__len__, __getitem__, __setitem__)

These dunder methods allow your custom classes to behave like native Python containers (lists, dictionaries).__len__(self): Enables the built-in len() function.  __getitem__(self, key): Enables indexing or key lookup syntax like obj[key].__setitem__(self, key, value): Enables item assignment syntax like obj[key] = value.

```python
class CustomShelf:

  def __init__(self, books):
    self.books = books

  def __len__(self):
    return len(self.books)

  def __getitem__(self, index):
    return self.books[index]

  def __setitem__(self, index, value):
    self.books[index] = value


shelf = CustomShelf(["Python 101", "Advanced Python"])
print(len(shelf))  # Output: 2 (calls __len__)
print(shelf[0])  # Output: Python 101 (calls __getitem__)
shelf[1] = "Expert Python"  # Calls __setitem__
```

# 3. Comparison & Hashing (__eq__, __lt__, __hash__)

By default, custom class instances are compared by their memory addresses. Dunder methods let you define custom equality, ordering rules, and hash values.

**\_\_eq\_\_(self, other)**: Defines behavior for the equality operator (==). 

**\_\_lt\_\_(self, other):** Defines behavior for the less-than operator (<). 

**\_\_hash\_\_(self):** Returns an integer hash value; required if you want your objects to work as dictionary keys or items inside a set.  

```python
class Temperature:

  def __init__(self, celsius):
    self.celsius = celsius

  def __eq__(self, other):
    return self.celsius == other.celsius

  def __lt__(self, other):
    return self.celsius < other.celsius

  def __hash__(self):
    return hash(self.celsius)


t1 = Temperature(25)
t2 = Temperature(25)
t3 = Temperature(30)

print(t1 == t2)  # Output: True (calls __eq__)
print(t1 < t3)  # Output: True (calls __lt__)
print({t1, t3})  # Works in sets because __hash__ is defined
```

# 4. Iteration & Execution (\_\_iter\_\_, \_\_next\_\_, \_\_call\_\_)
These methods hook your objects directly into Python's looping engines and execution flows.

\_\_iter\_\_ & \_\_next\_\_: Turns any custom class into a native iterator.

\_\_call\_\_(self, ...): Allows an instance of a class to be called like a regular function

```python
# 1. Callable Object Example
class Greeter:

  def __init__(self, greeting):
    self.greeting = greeting

  def __call__(self, name):
    return f"{self.greeting}, {name}!"


say_hello = Greeter("Hello")
print(say_hello("Alice"))  # Output: Hello, Alice! (Treats object as a function)


# 2. Custom Iterator Example
class Countdown:

  def __init__(self, start):
    self.current = start

  def __iter__(self):
    return self

  def __next__(self):
    if self.current <= 0:
      raise StopIteration
    val = self.current
    self.current -= 1
    return val


for num in Countdown(3):
  print(num, end=" ")  # Output: 3 2 1
```

# 5. Context Management & Resource Control (\_\_enter\_\_, \_\_exit\_\_)

These dunder methods transform your custom classes into context managers, empowering them to work inside with statements for clean setup and tear-down logic. 

\_\_enter\_\_(self): Executed at the beginning of the with block (returns the resource).

\_\_exit\_\_(self, exc_type, exc_val, exc_tb): Executed at the end of the with block, even if an exception crashes the inner code block.

```python
class ManagedResource:

  def __enter__(self):
    print("-> Resource acquired/opened.")
    return self

  def __exit__(self, exc_type, exc_val, exc_tb):
    print("-> Resource cleaned up/closed safely.")
    # Returning False lets exceptions propagate normally if they occurred


with ManagedResource():
  print("Inside the block: Doing work...")
# Output:
# -> Resource acquired/opened.
# Inside the block: Doing work...
# -> Resource cleaned up/closed safely.
```
