# Software Craftsmanship & Code Quality
Writing code that works is only half the battle. As a senior engineer, your code must be maintainable, testable, resilient, and easy for other developers to read. In this part, we examine the foundational pillars of software craftsmanship: SOLID Principles, Clean Code practices, Code Smells, Refactoring strategies, and Dependency Injection.

## 1. SOLID Principles
Popularized by Robert C. Martin ("Uncle Bob"), the SOLID principles are five design rules for writing maintainable object-oriented software:

S — Single Responsibility Principle (SRP): A class should have only one reason to change, meaning it should have only one job or responsibility.

O — Open/Closed Principle (OCP): Software entities (classes, modules, functions) should be open for extension, but closed for modification. You add new behavior by writing new code, not by changing old, tested code.

L — Liskov Substitution Principle (LSP): Objects of a superclass should be replaceable with objects of its subclasses without breaking the application's correctness.

I — Interface Segregation Principle (ISP): A client should never be forced to implement an interface it doesn't use, or be forced to depend on methods it doesn't need. (In Python, this translates to keeping protocols and base classes small and focused).

D — Dependency Inversion Principle (DIP): High-level modules should not depend on low-level modules; both should depend on abstractions. Abstractions should not depend on details.

## 2. Clean Code & Code Smells
Clean Code Practices:
Meaningful Names: Variable and function names should reveal intent (max_retry_attempts instead of mra).

Small Functions: Functions should do one thing, do it well, and fit comfortably on a single screen (ideally under 20 lines).

Comments as a Last Resort: Code should be self-documenting. Comments should explain why something is done, not what the code is doing.

Common Code Smells (Red Flags in Codebases):
Long Methods / Large Classes: A clear sign of a violation of the Single Responsibility Principle.

Duplicate Code (The DRY Principle Violation): Copy-pasting logic across modules instead of extracting it into a shared utility or base class.

Magic Numbers: Unexplained hardcoded values (e.g., if status == 4:). Replace them with named constants or enums.

Primitive Obsession: Using primitive data types (strings, integers) to represent complex domain concepts instead of dedicated small classes or dataclasses.

## 3. Refactoring
Refactoring is the disciplined process of restructuring existing computer code—changing the internal structure—without changing its external behavior.

When to Refactor: During the "Red-Green-Refactor" cycle of Test-Driven Development (TDD), or when preparing a codebase to accept new features.

Common Refactoring Techniques:

Extract Function/Method: Pulling a complex block of code out into its own descriptive function.

Replace Conditional with Polymorphism: Swapping massive if/elif/switch statements for object-oriented strategy classes.

Introduce Parameter Object: Grouping multiple primitive function arguments into a single cohesive configuration class or dataclass.

## 4. Dependency Injection (DI)
Dependency Injection ensures that classes do not instantiate their own dependencies internally. Instead, dependencies are passed (injected) into them from the outside (via constructors, properties, or method arguments).

```python
from abc import ABC, abstractmethod

class Logger(ABC):
    @abstractmethod
    def log(self, message: str): pass

class ConsoleLogger(Logger):
    def log(self, message: str):
        print(f"[LOG]: {message}")

class OrderProcessor:
    # Dependency is injected, making OrderProcessor completely decoupled and easy to unit test
    def __init__(self, logger: Logger):
        self.logger = logger

    def process_order(self, order_id: str):
        # Business logic...
        self.logger.log(f"Order {order_id} processed successfully.")
```

### Best Practices & Common Pitfalls
Test Before Refactoring: Never refactor complex legacy code unless you have a robust suite of automated unit tests safeguarding the existing behavior.


Avoid Premature Abstraction: Don't apply heavy design patterns or rigid interfaces before you actually need them. Keep it simple until requirements demand scalability.
