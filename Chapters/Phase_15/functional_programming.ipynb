# Functional Programming
Functional programming is a programming paradigm where computation is treated as the evaluation of mathematical functions and avoids changing-state and mutable data. While Python is primarily an object-oriented and imperative language, it borrows many powerful functional programming concepts that allow you to write concise, readable, and efficient code.

## 1. lambda Functions
A lambda function is a small, anonymous (unnamed) function defined using the lambda keyword. Unlike standard functions defined with def, lambda functions can only contain a single expression and automatically return its result.

### Syntax and Usage

```python
# Syntax: lambda arguments: expression

# Traditional function
def square(x):
    return x * x

# Equivalent lambda function
square_lambda = lambda x: x * x

print(square_lambda(5))  # Output: 25
```

Lambda functions are most commonly used as short, throwaway functions passed as arguments to higher-order functions (like map, filter, or sorting keys).

## 2. map()
The map() function applies a given function to all items in an iterable (like a list or tuple) and returns an iterator (a map object) containing the results.

```python
numbers = [1, 2, 3, 4, 5]

# Double every number using map and a lambda
doubled = list(map(lambda x: x * 2, numbers))

print(doubled)  # Output: [2, 4, 6, 8, 10]
```

Tip: List comprehensions ([x * 2 for x in numbers]) are often preferred over map() in Python for readability, but map() is useful when applying an existing function to multiple iterables simultaneously.

## 3. filter()
The filter() function constructs an iterator from elements of an iterable for which a given function returns True.

```python
numbers = [1, 2, 3, 4, 5, 6]

# Keep only even numbers
evens = list(filter(lambda x: x % 2 == 0, numbers))

print(evens)  # Output: [2, 4, 6]
```

## 4. reduce()
The reduce() function performs a rolling computation to sequential pairs of values in a list. It is part of the built-in functools module.

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

# Calculate the sum of all numbers
# Step-by-step: ((1 + 2) + 3) + 4) + 5
total = reduce(lambda x, y: x + y, numbers)

print(total)  # Output: 15
```

## 5. partial()
Also from the functools module, partial() allows you to "freeze" a portion of a function's arguments, creating a new function with fewer parameters.

```python
from functools import partial

def power(base, exponent):
    return base ** exponent

# Create a specialized function for squaring numbers
square = partial(power, exponent=2)

print(square(5))  # Output: 25 (5^2)
print(square(3))  # Output: 9  (3^2)
```

## 6. any() and all()
These are built-in functions that evaluate iterables of booleans and return a single boolean value.

any(iterable): Returns True if at least one element in the iterable is truthy. Returns False if all are falsy (or if the iterable is empty).

all(iterable): Returns True if all elements in the iterable are truthy. Returns False if even one element is falsy.

```python
status_flags = [False, True, False]

print(any(status_flags))  # Output: True (because at least one is True)
print(all(status_flags))  # Output: False (because not all are True)
```

## 7. sorted() and Key Functions
The sorted() function returns a new sorted list from the elements of any iterable. By default, it sorts items in ascending order.

```python
numbers = [5, 2, 9, 1]
print(sorted(numbers))  # Output: [1, 2, 5, 9]
```

## Key Functions (key=...)
Both sorted() and methods like max() and min() accept a key parameter. The key expects a function (often a lambda) that takes a single argument and returns a value to be used for sorting purposes instead of the original items.

```python
# A list of dictionaries representing students
students = [
    {"name": "Alice", "grade": 85},
    {"name": "Bob", "grade": 92},
    {"name": "Charlie", "grade": 78}
]

# Sort students by grade in descending order
sorted_students = sorted(students, key=lambda s: s["grade"], reverse=True)

print(sorted_students)
# Output: 
# [{'name': 'Bob', 'grade': 92}, {'name': 'Alice', 'grade': 85}, {'name': 'Charlie', 'grade': 78}]
```

### Best Practices & Common Pitfalls
**Favor Readability:** While functional tools like map and filter are powerful, Python developers often prefer list comprehensions or generator expressions because they are frequently easier to read and maintain.

**Remember reduce() Imports:** Unlike map and filter, reduce() is not built directly into the global namespace in Python 3; you must import it explicitly from functools.

**Be Careful with any() and all() on Generators:** Both any() and all() benefit from short-circuit evaluation (they stop evaluating as soon as the outcome is determined). Passing a generator expression to them is efficient because it won't evaluate items unnecessarily.
