# Behavioral Patterns & Dependency Injection
In this final part of Phase 27, we explore Behavioral patterns (which focus on communication and assignment of responsibilities between objects) and Dependency Injection (a core architectural principle for decoupling components and enabling modular testing).

## 1. Strategy Pattern
The Strategy pattern defines a family of algorithms, encapsulates each one into a separate class, and makes their objects interchangeable at runtime. This allows the algorithm to vary independently from the clients that use it.

```python
from abc import ABC, abstractmethod

class PaymentStrategy(ABC):
    @abstractmethod
    def pay(self, amount: float) -> str:
        pass

class CreditCardPayment(PaymentStrategy):
    def pay(self, amount: float) -> str:
        return f"Paid ${amount:.2f} using Credit Card."

class PayPalPayment(PaymentStrategy):
    def pay(self, amount: float) -> str:
        return f"Paid ${amount:.2f} using PayPal."

class ShoppingCart:
    def __init__(self, strategy: PaymentStrategy):
        self._strategy = strategy

    def set_strategy(self, strategy: PaymentStrategy):
        self._strategy = strategy

    def checkout(self, amount: float) -> str:
        return self._strategy.pay(amount)

# Usage
cart = ShoppingCart(CreditCardPayment())
print(cart.checkout(150.00))  # Paid $150.00 using Credit Card.

# Switch strategy dynamically at runtime
cart.set_strategy(PayPalPayment())
print(cart.checkout(75.50))   # Paid $75.50 using PayPal.
```

## 2. Observer Pattern
The Observer pattern defines a subscription mechanism to notify multiple objects ("observers") about any events happening to the object they are observing ("subject").

```python
class Subject:
    def __init__(self):
        self._observers = []
        self._state = None

    def attach(self, observer):
        self._observers.append(observer)

    def detach(self, observer):
        self._observers.remove(observer)

    def notify(self):
        for observer in self._observers:
            observer.update(self._state)

    def set_state(self, state: str):
        self._state = state
        self.notify()

class Subscriber:
    def __init__(self, name: str):
        self.name = name

    def update(self, state: str):
        print(f"{self.name} notified: System state changed to '{state}'")

# Usage
channel = Subject()
sub1 = Subscriber("Alice")
sub2 = Subscriber("Bob")

channel.attach(sub1)
channel.attach(sub2)

channel.set_state("Maintenance Mode")
# Output:
# Alice notified: System state changed to 'Maintenance Mode'
# Bob notified: System state changed to 'Maintenance Mode'
```

## 3. Dependency Injection (DI)
Dependency Injection is an architectural design pattern where an object receives its dependencies from an external source rather than creating them internally. It drastically improves code modularity, flexibility, and testability.

```python
class EmailService:
    def send_email(self, message: str):
        print(f"Sending email: {message}")

class NotificationManager:
    # Dependency is injected via constructor (Constructor Injection)
    def __init__(self, email_service: EmailService):
        self.email_service = email_service

    def alert(self, message: str):
        self.email_service.send_email(message)

# Usage
service = EmailService()
manager = NotificationManager(service)  # Injecting the dependency
manager.alert("Server CPU usage is above 90%!")
```

### Best Practices & Common Pitfalls
**Favor Composition over Inheritance:** Patterns like Strategy and Dependency Injection rely heavily on passing objects into classes rather than subclassing, keeping your codebase flexible and preventing deep, brittle class hierarchies.

**Avoid Overcomplicating Simple Code:** If your script only requires a single straightforward function or condition, don't build full Strategy or Observer class structures. Apply patterns when behaviors are expected to scale or change over time.
