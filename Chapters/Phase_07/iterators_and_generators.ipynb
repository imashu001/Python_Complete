# 1. Iterable vs. Iterator
Before diving into code, it is critical to distinguish between these two fundamental terms:

**Iterable:** An object that you can loop over. Examples include lists, tuples, strings, dictionaries, and sets. An iterable has an __iter__() method that can produce an iterator, but it is not an iterator itself (it doesn't keep track of a current state or have a __next__() method). 

**Iterator:** An object that represents a stream of data and keeps track of its current state. It remembers where it is in a sequence. When you call next() on it, it yields the next item. Once exhausted, it raises a StopIteration exception.  

# 2. iter() and next()
Every time you write a for loop in Python, Python automatically calls the iter() function on your collection under the hood to fetch an iterator, and then repeatedly calls next() until a StopIteration exception is raised.

You can also use these built-in functions manually:

```python
# A list is an iterable, but NOT an iterator
numbers = [10, 20, 30]

# Fetching the iterator using iter()
iterator = iter(numbers)

# Manually pulling items using next()
print(next(iterator))  # Output: 10
print(next(iterator))  # Output: 20
print(next(iterator))  # Output: 30

# Calling next() again will trigger StopIteration
# print(next(iterator)) # Raises StopIteration error
```

# 3. Generator Functions & yield

Writing custom iterator classes using __iter__() and __next__() requires boilerplate code. Python solves this elegantly with generators.

A generator function looks like a normal function, but instead of using return to exit permanently, it uses the yield keyword to emit a value and pause execution.  

**How yield Works**
When yield is hit, the function returns the value and freezes its current state (local variables, instruction pointer).  The next time next() is called on the generator, it wakes up right where it left off and continues running until the next yield (or until the function terminates).  

```python
def countdown(n):
  print("Starting countdown...")
  while n > 0:
    yield n
    n -= 1
  print("Countdown finished!")


# Creating the generator object
gen = countdown(3)

print(next(gen))  # Resumes, prints "Starting countdown...", yields 3
print(next(gen))  # Resumes loop, yields 2
print(next(gen))  # Resumes loop, yields 1
# print(next(gen)) # Raises StopIteration (and prints "Countdown finished!")
```

# 4. Generator Expressions
As you saw briefly in comprehensions, you can create anonymous generators instantly using generator expressions by replacing square brackets [] with parentheses ().

```python
# Generator expression for squares up to 1000000
squares_gen = (x**2 for x in range(1_000_000))

print(type(squares_gen))  # Output: <class 'generator'>
print(next(squares_gen))  # Output: 0
print(next(squares_gen))  # Output: 1
```

# 5. Lazy Evaluation
The core superpower of iterators and generators is lazy evaluation—an evaluation strategy where values are computed only on demand, right when they are requested.  

**Why Lazy Evaluation Matters:** 
**Memory Efficiency:** If you need to process a 10GB log file or calculate a sequence of a billion items, a list comprehension will crash your computer by trying to load everything into RAM simultaneously. A generator keeps only one item in memory at a time. 

**Infinite Streams:** You can build infinite data pipelines (like a continuous stream of sensor data, or an infinite Fibonacci sequence) because computation stops when you stop asking for elements.

```python
# An infinite sequence generator
def infinite_evens():
  num = 0
  while True:
    yield num
    num += 2


even_gen = infinite_evens()

# Pulling just the first few values safely without crashing memory
for _ in range(5):
  print(next(even_gen), end=" ")  # Output: 0 2 4 6 8
```
