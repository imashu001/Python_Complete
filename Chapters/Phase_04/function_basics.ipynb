# Phase 4A: Function Basics, Arguments, & Scope

A **function** is a reusable block of code designed to perform a specific task. Instead of repeating code, you define it once inside a function and call it whenever you need it.

---

## 1. Defining and Calling Functions

Functions are defined using the `def` keyword, followed by the function name, parentheses `()`, and a colon `:`

```python
# Defining a function
def greet_user():
    print("Hello, welcome to Python!")


# Calling (executing) the function
greet_user()

## 2. Parameters vs Arguments
Parameters: The variables listed inside the parentheses when defining a function.

Arguments: The actual values passed to the function when it is called.

````python
# 'name' is a parameter
def say_hello(name):
    print(f"Hello, {name}!")


# "Alice" is the argument passed in
say_hello("Alice")
````

## 3. Types of Arguments
Python offers incredible flexibility in how you pass arguments to functions.

Positional Arguments
Arguments passed in the exact order the parameters are defined.

````python
def describe_pet(animal_type, pet_name):
    print(f"I have a {animal_type} named {pet_name}.")


describe_pet("dog", "Buddy")  # Order matters!
````

Keyword Arguments
Arguments passed with explicit name=value pairs, meaning order no longer matters.

````python
describe_pet(pet_name="Buddy", animal_type="dog")
````

Default Arguments
Parameters can be given default values. If no argument is passed for that parameter during the call, the default value is used.

````python
def describe_pet(pet_name, animal_type="dog"):
    print(f"I have a {animal_type} named {pet_name}.")


describe_pet("Max")  # Uses default "dog" for animal_type
````

Variable-Length Arguments (*args)
Used to pass a non-keyword variable number of positional arguments. Python packs them into a tuple.

````python
def sum_all(*args):
    return sum(args)


print(sum_all(1, 2, 3, 4))  # Output: 10
````

Keyword Variable-Length Arguments (**kwargs)
Used to pass a variable number of keyword arguments. Python packs them into a dictionary.

````python
def print_user_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")


print_user_info(name="Alice", age=30, city="New York")
````

## 4. Return Values
Functions can send data back to the caller using the return keyword. If a function has no return statement, it implicitly returns None.

````python
def add(a, b):
    return a + b


result = add(5, 3)
print(result)  # Output: 8
````

You can also return multiple values (which are packed and returned as a tuple):

````python
def get_min_max(numbers):
    return min(numbers), max(numbers)


lowest, highest = get_min_max([10, 2, 8, 45])
````

## 5. Scope (The LEGB Rule)
Scope determines the visibility and lifetime of a variable within your program. Python checks scopes using the LEGB rule (from innermost to outermost):

L - Local: Names assigned inside a current function.

E - Enclosing: Names in the local scope of any enclosing (nested) functions.

G - Global: Names assigned at the top-level of a module file or declared global.

B - Built-in: Names preassigned in Python's built-in module (e.g., print, len).

````python
x = "Global Variable"  # Global Scope


def outer():
    x = "Enclosing Variable"  # Enclosing Scope

    def inner():
        x = "Local Variable"  # Local Scope
        print(x)

    inner()


outer()
````
