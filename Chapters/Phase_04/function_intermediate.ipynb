## 1. Lambda (Anonymous) Functions

A **lambda** function is a small, anonymous (nameless) function defined using the `lambda` keyword. They can take any number of arguments, but can only contain **a single expression**. 

Lambda functions are typically used for short, throwaway operations (often inside higher-order functions like `map()` or `sorted()`).

```python
# Regular function
def square(x):
  return x * x


# Equivalent lambda function
square_lambda = lambda x: x * x

print(square_lambda(5))  # Output: 25

# Commonly used with sorted()
students = [("Alice", 85), ("Bob", 92), ("Charlie", 78)]
# Sort students by their score (second element in tuple)
students.sort(key=lambda student: student[1])
print(students)  # Output: [('Charlie', 78), ('Alice', 85), ('Bob', 92)]

## 2. Recursion
Recursion occurs when a function calls itself to solve a smaller instance of the same problem. Every recursive function must have a base case to prevent infinite recursion and a resulting RecursionError.

````python
def factorial(n):
  # Base case: stop recursion when n is 0 or 1
  if n == 0 or n == 1:
    return 1
  # Recursive step
  return n * factorial(n - 1)


print(factorial(5))  # Output: 120 (5 * 4 * 3 * 2 * 1)
````

## 3. Nested Functions & Scope Modification
You can define functions inside other functions. These are called nested functions or inner functions.

````python
def outer_function(text):
  # Inner function can access variables of the outer function
  def inner_function():
    print(text)

  inner_function()


outer_function("Hello from the inner function!")
````

The nonlocal Keyword
By default, if you assign a value to a variable inside an inner function, Python creates a brand-new local variable. If you want to modify a variable belonging to the enclosing (outer) function, you must use the nonlocal keyword.

````python
def outer():
  count = 0

  def inner():
    nonlocal count
    count += 1
    return count

  return inner


counter = outer()
print(counter())  # Output: 1
print(counter())  # Output: 2
````

## 4. Closures
A closure is a nested function that remembers and has access to variables in the enclosing scope even after the outer function has finished executing and returned.

Closures are useful for data hiding and creating customized function factories.

````python
def multiplier_factory(factor):
  # 'factor' is enclosed by the inner function
  def multiply(x):
    return x * factor

  return multiply


# Create specialized functions
double = multiplier_factory(2)
triple = multiplier_factory(3)

print(double(5))  # Output: 10
print(triple(5))  # Output: 15
````

## 5. First-Class Functions
In Python, functions are first-class objects. This means functions can be treated just like any other object (like integers, strings, or lists):

Assigned to variables.

Stored in data structures (lists, dictionaries).

Passed as arguments to other functions.

Returned from other functions.

````python
def yell(text):
  return text.upper() + "!"


def whisper(text):
  return text.lower() + "..."


# Storing functions in a data structure
actions = {"shout": yell, "murmur": whisper}

print(actions["shout"]("hello"))  # Output: HELLO!
````

## 6. Higher-Order Functions
A higher-order function is a function that either:

Takes one or more functions as arguments, OR

Returns a function as its result.

Because Python treats functions as first-class objects, higher-order functions are built right into the language's design.

````python
# A higher-order function that accepts another function as an argument
def apply_operation(func, value):
  return func(value)


result = apply_operation(lambda x: x ** 3, 3)
print(result)  # Output: 27 (3 cubed)
````
