###  Golden Rule of Ordering
When catching multiple exceptions, always place more specific child exceptions before broader parent exceptions. If you place a broad exception (like Exception) first, it will intercept everything, rendering your specific sub-handlers useless.  

```python
try:
  # Some risky operations
  result = 10 / int("0")
except ZeroDivisionError:
  print("Caught a specific error: Division by zero!")
except Exception as e:
  print(f"Caught a general standard exception: {e}")
```

# 2. else and finally in Error Handling
A complete try block can be extended with two additional clauses: else and finally.

else: Runs only if the try block executes without raising any exceptions. It keeps your try block clean by isolating code that should only run on success.

finally: Runs unconditionally—no matter what. Whether an exception occurred, was caught, or the code returned early, finally will execute (often used for cleanup tasks like closing database connections or files).

```python
def process_data(user_input):
  try:
    value = int(user_input)
    result = 100 / value
  except (ValueError, ZeroDivisionError) as e:
    print(f"Error occurred: {e}")
  else:
    # Runs ONLY if no exception happened inside the try block
    print(f"Success! Result is {result}")
  finally:
    # Always runs as a cleanup step
    print("Execution complete. Cleaning up resources...\n")


process_data("10")  # Triggers else and finally
process_data("abc")  # Triggers except and finally
```

# 3. The raise Statement
You are not limited to catching exceptions thrown by Python; you can actively trigger exceptions yourself using the raise keyword. This is useful when your code encounters a state that violates business logic.

```python
def set_age(age):
  if age < 0 or age > 120:
    raise ValueError("Age must be realistic between 0 and 120.")
  print(f"Age set to {age}")


try:
  set_age(150)
except ValueError as err:
  print(f"Blocked invalid input: {err}")
```

# 4. Custom Exceptions
When building larger applications or libraries, relying solely on generic errors like ValueError can make debugging difficult. You can create custom exceptions by inheriting from Python's built-in Exception class.

By convention, custom exception names should end with the word Error.

```python
# 1. Define the custom exception class
class InsufficientFundsError(Exception):
  """Raised when an account has insufficient funds for a withdrawal."""

  def __init__(self, balance, amount_requested):
    self.balance = balance
    self.amount_requested = amount_requested
    super().__init__(
        f"Attempted to withdraw ${amount_requested}, but balance is only"
        f" ${balance}."
    )


# 2. Use the custom exception in your logic
class BankAccount:

  def __init__(self, balance):
    self.balance = balance

  def withdraw(self, amount):
    if amount > self.balance:
      raise InsufficientFundsError(self.balance, amount)
    self.balance -= amount
    return self.balance


# 3. Handling the custom exception
account = BankAccount(50)

try:
  account.withdraw(100)
except InsufficientFundsError as e:
  print(f"Transaction Failed: {e}")
  # You can access custom attributes safely!
  print(f"Shortfall amount: ${e.amount_requested - e.balance}")
```
