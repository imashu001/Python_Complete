## 1. Classes, Objects, and Constructors (__init__ & self)
**Class:** A blueprint or template for creating objects. It defines a set of attributes and methods that the created objects will have.

**Object (Instance):** An individual realization of a class.

**Constructor (__init__):** A special method that automatically runs when you create a new object from a class. It is used to initialize the object's starting attributes.

self: A reference to the specific object currently being created or interacted with. It allows each object to keep track of its own unique data.

```python
class Dog:
  # Constructor method to initialize attributes
  def __init__(self, name, breed):
    self.name = name  # Instance variable
    self.breed = breed  # Instance variable

  # Instance method
  def bark(self):
    return f"{self.name} says Woof!"


# Creating objects (instantiation)
dog1 = Dog("Buddy", "Golden Retriever")
dog2 = Dog("Max", "German Shepherd")

print(dog1.name)  # Output: Buddy
print(dog2.bark())  # Output: Max says Woof!
```

## 2. Instance Variables vs. Class Variables
Where you define your variables inside a class dictates their scope and sharing behavior:

**Instance Variables:** Tied to a specific instance (object). Every object gets its own separate copy of these variables (defined inside __init__ using self).

**Class Variables:** Shared across all instances of a class. They are defined directly inside the class body outside of any met

```python
class Employee:
  # Class variable (shared by all employees)
  company_name = "TechCorp"
  raise_amount = 1.05

  def __init__(self, name, salary):
    self.name = name  # Instance variable
    self.salary = salary  # Instance variable


emp1 = Employee("Alice", 50000)
emp2 = Employee("Bob", 60000)

print(emp1.company_name)  # Output: TechCorp
print(emp2.company_name)  # Output: TechCorp

# Changing a class variable affects all instances unless overridden
Employee.company_name = "InnoSoft"
print(emp1.company_name)  # Output: InnoSoft
```

## 3. Instance, Class, and Static Methods
Python classes support three distinct types of methods, each serving a different purpose regarding data access.

### A. Instance Methods
What they do: Operate on individual instances of the class.

First Parameter: self (accesses/modifies instance state).

### B. Class Methods
What they do: Operate on the class as a whole rather than a specific object. Often used as alternative constructors.

Decorator: @classmethod

First Parameter: cls (accesses/modifies class-level state).

### C. Static Methods
What they do: Utility functions that do not access or modify instance or class state. They behave like normal functions, but live inside the class namespace for logical grouping.

Decorator: @staticmethod

Parameters: Neither self nor cls.

```python
class MathOperations:
  pi = 3.14159  # Class variable

  def __init__(self, value):
    self.value = value

  # 1. Instance Method (uses self)
  fi
  def double_value(self):
    return self.value * 2

  # 2. Class Method (uses cls)
  @classmethod
  def get_pi(cls):
    return cls.pi

  # 3. Static Method (independent utility)
  @staticmethod
  def add(a, b):
    return a + b


# Usage examples:
op = MathOperations(10)
print(op.double_value())  # Instance method: 20
print(MathOperations.get_pi())  # Class method: 3.14159
print(MathOperations.add(5, 7))  # Static method: 12
```
