# 1. Inheritance & Method Overriding
Inheritance allows a new class (the child or subclass) to adopt the attributes and methods of an existing class (the parent or superclass). This promotes code reuse and establishes a natural "is-a" relationship (e.g., a Dog is an Animal).

Method Overriding: A subclass can redefine a method inherited from its parent class to provide behavior specific to the child.

```python
# Parent Class
class Animal:

  def __init__(self, name):
    self.name = name

  def make_sound(self):
    return "Some generic animal sound"


# Child Class inheriting from Animal
class Cat(Animal):

  # Overriding the parent's make_sound method
  def make_sound(self):
    return f"{self.name} says Meow!"


my_cat = Cat("Whiskers")
print(my_cat.name)  # Output: Whiskers (inherited from Animal)
print(my_cat.make_sound())  # Output: Whiskers says Meow! (overridden)
```

# 2. The super() Function
When you override methods (especially the __init__ constructor), you often still want to execute parts of the parent class's logic. The super() function allows you to temporarily access the parent class and call its methods.

```python
class Person:

  def __init__(self, name, age):
    self.name = name
    self.age = age


class Employee(Person):

  def __init__(self, name, age, employee_id):
    # Call the parent class's __init__ to handle name and age
    super().__init__(name, age)
    self.employee_id = employee_id

  def get_details(self):
    return f"Employee {self.name}, ID: {self.employee_id}"


emp = Employee("Alice", 30, "E1042")
print(emp.name)  # Output: Alice
print(emp.age)  # Output: 30
```

# 3. Multiple Inheritance
Python allows a class to inherit from more than one parent class simultaneously. While powerful, it introduces potential complexity if multiple parents define methods with the same name.

```python
class Flyer:

  def move(self):
    return "Flying high in the sky!"


class Swimmer:

  def move(self):
    return "Swimming through the water!"


# Multiple inheritance
class Duck(Flyer, Swimmer):
  pass


donald = Duck()
print(
    donald.move()
)  # Output: Flying high in the sky! (Resolves to the first parent listed)
```

# 4. Method Resolution Order (MRO)
When dealing with complex or multiple inheritance, how does Python decide which parent's method to run first? It uses the Method Resolution Order (MRO).

Python uses the C3 Linearization algorithm to determine a strict, predictable linear order in which base classes are searched.

You can inspect a class's MRO using the .mro() method or the __mro__ attribute.

```python
class A:

  def process(self):
    print("Process in A")


class B(A):

  def process(self):
    print("Process in B")


class C(A):

  def process(self):
    print("Process in C")


class D(B, C):
  pass


# Inspecting the Method Resolution Order
print(D.mro())
# Output order: [D, B, C, A, object]

d = D()
d.process()  # Output: Process in B (since B comes before C in D's inheritance list)
```
