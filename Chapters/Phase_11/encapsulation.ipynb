# 1. Encapsulation
Encapsulation is the practice of bundling data and the methods that operate on that data within a single unit (the class), while restricting direct access to some of an object's components. This protects internal state from unauthorized modification.

In Python, encapsulation is achieved through naming conventions rather than strict compiler enforcement:

Public: Standard attributes (e.g., self.name). Accessible anywhere.

Protected (Convention): Prefixed with a single underscore (e.g., self._balance). Indicates that the attribute should only be accessed within the class and its subclasses.

Private (Name Mangling): Prefixed with a double underscore (e.g., self.__pin). Python automatically renames this internally to prevent external access.

Getters and Setters (@property): Modern Python uses the @property decorator to manage access to private data safely, allowing validation logic during updates.

```python
class BankAccount:

  def __init__(self, owner, initial_balance):
    self.owner = owner
    self.__balance = initial_balance  # Private attribute

  # Getter property to safely read the private balance
  @property
  def balance(self):
    return self.__balance

  # Setter property to control how the balance is modified
  @balance.setter
  def balance(self, amount):
    if amount < 0:
      raise ValueError("Balance cannot be negative.")
    self.__balance = amount


acc = BankAccount("Alice", 1000)
print(acc.balance)  # Output: 1000 (calls the getter property)

acc.balance = 1500  # Calls the setter property
# acc.balance = -500  # Raises ValueError: Balance cannot be negative.
```

# 2. Abstraction & Abstract Base Classes (ABCs)
Abstraction hides complex implementation details and exposes only the essential features. In Python, this is implemented using Abstract Base Classes (ABCs) via the built-in abc module.

An abstract class cannot be instantiated directly.

It serves as a blueprint or contract, forcing subclasses to implement specific methods using the @abstractmethod decorator.

```python
from abc import ABC, abstractmethod


# Abstract Base Class
class PaymentProcessor(ABC):

  @abstractmethod
  def process_payment(self, amount):
    """Subclasses must implement this method"""
    pass


# Concrete Subclass
class CreditCardProcessor(PaymentProcessor):

  def process_payment(self, amount):
    return f"Processing ${amount} via Credit Card."


# processor = PaymentProcessor() # TypeError: Can't instantiate abstract class
processor = CreditCardProcessor()
print(processor.process_payment(100))  # Output: Processing $100 via Credit Card.
```

# 3. Polymorphism
Polymorphism means "many forms." It allows different classes to have methods with the same name, enabling functions or loops to treat different types of objects uniformly.

Python achieves polymorphism naturally through its dynamic nature (often called Duck Typing: "If it walks like a duck and quacks like a duck, it's a duck"), or explicitly through inheritance.

```python
class Dog:

  def make_sound(self):
    return "Woof!"


class Cat:

  def make_sound(self):
    return "Meow!"


# A polymorphic function that accepts any object with a 'make_sound' method
def animal_concert(animal_object):
  print(animal_object.make_sound())


animal_concert(Dog())  # Output: Woof!
animal_concert(Cat())  # Output: Meow!
```

# 4. Composition vs. Inheritance ("Has-a" vs. "Is-a")
One of the most important architectural decisions in OOP is choosing between inheritance and composition.

Inheritance ("Is-a" relationship): A SportsCar is a Car. Use this only when there is a true hierarchical relationship and code sharing makes sense. Overusing inheritance leads to rigid, fragile codebases.

Composition ("Has-a" relationship): A Car has an Engine, has a GPS, and has Wheels. Instead of inheriting behaviors, a class contains instances of other classes as attributes.

Best Practice: Favor composition over inheritance whenever possible. It keeps classes loosely coupled, highly modular, and much easier to test and modify.

```python
# Component classes
class Engine:

  def start(self):
    return "Engine started Vroom!"


class GPS:

  def navigate(self):
    return "Navigating to destination..."


# Composition: Car 'has-a' Engine and 'has-a' GPS
class Car:

  def __init__(self, make, model):
    self.make = make
    self.model = model
    self.engine = Engine()  # Composed object
    self.gps = GPS()  # Composed object

  def drive(self):
    print(self.engine.start())
    print(self.gps.navigate())


my_car = Car("Toyota", "Camry")
my_car.drive()
# Output:
# Engine started Vroom!
# Navigating to destination...
```
