# Phase 3: Control Flow

By default, Python executes your code sequentially from top to bottom. Control flow statements allow you to alter this execution path by making decisions (conditionals) or repeating blocks of code (loops).

---

## 1. Conditional Statements

Conditional statements allow your program to execute specific blocks of code only when certain conditions are met.

### `if` Statement
Executes a block of code if a condition evaluates to `True`.
```python
age = 18
if age >= 18:
    print("You are an adult.")

if-else Statement
Provides an alternative block of code to run if the if condition evaluates to False.

Python

````python
age = 16
if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
````

if-elif-else Statement
Allows you to check multiple conditions sequentially. elif stands for "else if".

Python

````python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: F")
````

Nested if Statements
Placing an if statement inside another if statement to evaluate deeper, multi-layered conditions.

````python
is_logged_in = True
is_admin = False

if is_logged_in:
    if is_admin:
        print("Welcome to the admin dashboard.")
    else:
    print("Welcome to the user dashboard.")
else:
    print("Please log in first.")
````

match-case Statement (Python 3.10+)
Structural pattern matching that acts like a powerful, cleaner alternative to long if-elif-else chains.

````python
status_code = 200

match status_code:
    case 200:
        print("Success")
    case 400 | 404:
        print("Client Error / Not Found")
    case 500:
        print("Server Error")
    case _:
        print("Unknown status")
````

## 2. Looping Statements
Loops allow you to execute a block of code multiple times.

for Loops
Used for iterating over a sequence (such as a list, tuple, dictionary, set, or string).

````python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
````

while Loops
Repeats a block of code as long as a given condition evaluates to True.

````python
count = 3
while count > 0:
    print(f"Countdown: {count}")
    count -= 1
print("Blast off!")
````

## 3. Loop Control Statements
Keywords used to modify or control the behavior of loops.

break: Completely exits/terminates the loop immediately.

continue: Skips the current iteration and jumps straight to the next iteration of the loop.

pass: A null operation placeholder that does nothing. Used when syntax requires a statement, but you don't want any code to execute yet.

````python
# break and continue example
for num in range(1, 6):
    if num == 3:
        continue  # Skips printing 3
    if num == 5:
        break  # Stops the loop entirely when reaching 5
    print(num)
# Output: 1, 2, 4
````

## 4. Advanced Loop Utilities
Built-in utilities that make looping cleaner and more pythonic.

enumerate()
Adds a counter/index to an iterable, returning pairs of (index, element) during iteration.

Python

````python
fruits = ["apple", "banana", "cherry"]

for index, fruit in enumerate(fruits):
    print(f"Index {index}: {fruit}")

# Output:
# Index 0: apple
# Index 1: banana
# Index 2: cherry
````

zip()
Combines multiple iterables (lists, tuples, etc.) element-by-element, stopping when the shortest iterable is exhausted.

````python
names = ["Alice", "Bob", "Charlie"]
scores = [85, 92, 78]

for name, score in zip(names, scores):
    print(f"{name} scored {score}")

# Output:
# Alice scored 85
# Bob scored 92
# Charlie scored 78
````
