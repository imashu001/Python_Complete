# 1. is vs ==
== (Value Equality): Compares the values of two objects to see if they are equal. It invokes the __eq__() magic method under the hood.

is (Identity Comparison): Compares the memory addresses (id()) of two objects to check if they point to the exact same object in RAM.

```python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)  # True (values are identical)
print(a is b)  # False (different objects in memory)
```

# 2. Mutable vs Immutable Objects

Mutable objects: Can be modified in place after creation without changing their memory address. Examples: list, dict, set.

Immutable objects: Cannot be modified after creation. Any modification creates a brand-new object in memory. Examples: int, float, str, tuple, frozenset.

⚠️ Classic Interview Trap: Passing mutable default arguments in functions causes state persistence across function calls because the default object is evaluated only once when the function is defined.

# 3. Deep Copy vs Shallow Copy

Shallow Copy (copy.copy()): Creates a new container object, but populates it with references to the nested child objects found in the original.

Deep Copy (copy.deepcopy()): Recursively creates brand-new copies of the container and all its nested objects, ensuring zero memory sharing between the original and the copy.

# 4. List vs Tuple
list: A mutable sequence intended for homogeneous data collections whose size or contents may change over time. Uses dynamic arrays.

tuple: An immutable sequence intended for heterogeneous data structures. Because they are immutable, tuples are hashable (can be used as dictionary keys) and consume less memory than lists.

# 5. Set vs Dict

set: An unordered collection of unique, hashable elements implemented internally as a hash table storing only keys. Optimized for membership tests ($O(1)$ average time complexity).

dict: A collection of key-value pairs where keys must be unique and hashable.

# 6. Generator vs Iterator

Iterator: An object that implements the iterator protocol (\_\_iter\_\_() and \_\_next\_\_()). It traverses a container and raises StopIteration when exhausted.

Generator: A specialized, memory-efficient type of iterator created using functions with the yield keyword or generator expressions. It computes items on the fly, keeping memory footprint at $O(1)$.

# 7. Thread vs Process
Process: An independent execution instance of a program with its own private memory space and Python interpreter. Bypasses the GIL (true parallelism).

Thread: A lightweight execution unit sharing the same memory space within a single process. Subject to the Global Interpreter Lock (GIL) in CPython.

# 8. async vs threading

threading (Preemptive Concurrency): Managed by the operating system kernel. Threads switch unpredictably based on time slices, requiring locks (Mutex) to avoid race conditions.

asyncio (Cooperative Concurrency): Single-threaded event loop. Tasks explicitly yield control back to the event loop using await when waiting for I/O operations, eliminating context-switch overhead.


# 9. @staticmethod vs @classmethod

@staticmethod: A regular function living inside a class namespace. It takes no implicit first argument (self or cls) and cannot access instance or class state.

@classmethod: Takes the class itself (cls) as its first implicit argument. Commonly used as alternative constructors.

```python
class DateTime:
    def __init__(self, year, month, day):
        self.year, self.month, self.day = year, month, day

    @classmethod
    def from_string(cls, date_str):
        year, month, day = map(int, date_str.split("-"))
        return cls(year, month, day)  # Calls constructor dynamically
```

# 10. LEGB Rule (Variable Scope Resolution)
Python resolves variable names using the LEGB order:

L — Local: Inside the current function or method.

E — Enclosing: Any surrounding functions (closures).

G — Global: At the module/script level.

B — Built-in: Python's pre-defined names (print, len, int).

# 11. GIL (Global Interpreter Lock)
A mutual exclusion lock in CPython ensuring that only one native thread executes Python bytecode at a time, preventing multi-threaded CPU-bound speedups. Solved using multiprocessing for CPU workloads and asyncio/threading for I/O workloads.

# 12. MRO (Method Resolution Order) & C3 Linearization
The order in which Python searches for inherited methods in complex or multiple-inheritance hierarchies. You can inspect any class's MRO using ClassName.__mro__ or ClassName.mro(). Python uses the C3 Linearization algorithm to guarantee monotonicity and local precedence ordering.

# 13. Monkey Patching
Dynamically modifying or overriding modules, classes, or functions at runtime without altering the original source code. While powerful for mocking tests or hot-fixing legacy codebases, it can make code logic extremely difficult to debug.

# 14. Metaclasses
A metaclass is "a class of a class." Just as a class defines how an instance behaves, a metaclass defines how a class behaves. Created by inheriting from type, they are used in ORMs (like Django Models) to intercept and modify class creation at definition time.

```python
class Meta(type):
    def __new__(cls, name, bases, attrs):
        attrs['custom_attribute'] = 'Injected by Metaclass'
        return super().__new__(cls, name, bases, attrs)

class MyClass(metaclass=Meta):
    pass

print(MyClass.custom_attribute)  # Injected by Metaclass
```

# 15. Context Managers & \_\_enter\_\_ / \_\_exit\_\_
Objects that manage resource allocation and cleanup (commonly seen with the with statement). They must implement two magic methods:

\_\_enter\_\_(self): Sets up the resource and returns the target object.

\_\_exit\_\_(self, exc_type, exc_val, exc_tb): Guarantees cleanup execution, even if exceptions occur inside the block.
